### What is Ensemble Learning?
**Ensemble Learning** = combining **multiple models** together to make BETTER predictions than any single model alone!

## What is it?

**Real-life analogy:** Imagine you have a difficult math problem.
- Asking **ONE friend** for the answer → they might be wrong.
- Asking **5 different friends** and taking the majority answer → much more reliable!
- Even better — if you know which friends are BEST at math, give their answers more weight!

That's exactly what Ensemble Learning does! It's the **"wisdom of the crowd"** applied to Machine Learning. 🧠✨

### The Golden Rule of Ensembles:
> **"Many weak learners together > One strong learner alone"**

---

## 🔑 The 3 Main Types of Ensemble Learning

### 1️⃣ **Bagging (Bootstrap Aggregating)** 🎒
- Train MANY copies of the SAME algorithm on DIFFERENT random subsets of data.
- Each model votes → majority wins.
- **Example:** Random Forest = Bagging with Decision Trees!
- **Reduces:** Variance (overfitting).

### 2️⃣ **Boosting** 🚀
- Train models SEQUENTIALLY — each new model fixes previous mistakes.
- **Examples:** AdaBoost, Gradient Boosting, XGBoost (we've done all 3!)
- **Reduces:** Bias (underfitting).

### 3️⃣ **Voting / Stacking** 🗳️
- Combine DIFFERENT algorithms (e.g., Logistic Regression + Random Forest + SVM).
- **Voting:** Simple majority or weighted vote.
- **Stacking:** A "meta-model" learns how to best combine the predictions.
- **Reduces:** Both bias AND variance!

---

## 🔑 Voting Types

| Type | How it works | When to use |
|---|---|---|
| **Hard Voting** | Each model gives a class prediction → majority wins | When models don't give probabilities |
| **Soft Voting** | Each model gives probabilities → average them → highest wins | ✅ **Usually better!** |

**Example (Hard Voting):**
- Model 1: Diabetes ✅
- Model 2: Diabetes ✅
- Model 3: No Diabetes ❌
- Result: **Diabetes** (2 vs 1)

**Example (Soft Voting):**
- Model 1: 80% Diabetes
- Model 2: 60% Diabetes
- Model 3: 40% Diabetes
- Average: (80+60+40)/3 = **60% → Diabetes**

````python
# ================================================================
# STEP 1: IMPORT LIBRARIES
# ================================================================

# --- Core data tools ---
import numpy as np
import pandas as pd

# --- Plotting tools ---
import matplotlib.pyplot as plt
import seaborn as sns

# --- Saving the trained model ---
import joblib

# --- Splitting data ---
from sklearn.model_selection import train_test_split, cross_val_score

# --- Feature scaling ---
from sklearn.preprocessing import StandardScaler

# --- Base models for our ensemble ---
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB

# --- 🌟 The Ensemble Classes — the stars of this notebook! ---
from sklearn.ensemble import (VotingClassifier,
                               BaggingClassifier,
                               StackingClassifier,
                               RandomForestClassifier)

# --- Classification metrics ---
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                              confusion_matrix, classification_report,
                              roc_auc_score, roc_curve)

# --- Make charts look nice ---
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 12

# --- Hide harmless warning messages ---
import warnings
warnings.filterwarnings('ignore')

print("✅ All necessary libraries imported successfully!")
````

## Step 2 — Load the Dataset

Same as the Logistic Regression notebook so we can compare!

- Each row = one patient.
- `Outcome` column = 1 if diabetic, 0 if not (our TARGET).

````python
# Load the diabetes dataset from your directory
df = pd.read_csv("diabetes.csv")

print(f"Total Patients (Rows): {df.shape[0]}")
print(f"Total Columns         : {df.shape[1]}")
df.head()
````

# Step 3 — Quick EDA Recap

We already explored this dataset in the Logistic Regression notebook. Here's a quick check:

````python
# ⚠️ Check target balance
print("Outcome counts:")
print(df['Outcome'].value_counts())
print("\nOutcome percentages:")
print((df['Outcome'].value_counts(normalize=True) * 100).round(1))

plt.figure(figsize=(6, 4))
sns.countplot(x='Outcome', data=df, palette='Set2')
plt.title("Diabetes Outcome Distribution")
plt.xlabel("Outcome (0 = No Diabetes, 1 = Diabetes)")
plt.show()
````

# Step 4 — Data Preprocessing (Fix the Hidden Missing Values!)

### 🚨 Reminder about this dataset:
Values of **0** for medical measurements like Glucose, BloodPressure, BMI are **biologically impossible!**
These are missing values in disguise → we replace with median.

````python
# ----------------------------------------------------------------
# Fix the hidden missing values (impossible 0s)
# ----------------------------------------------------------------
invalid_zero_cols = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

print("Impossible 0s BEFORE cleaning:")
for col in invalid_zero_cols:
    print(f"  {col:20s}: {(df[col] == 0).sum()} zeros")

# Replace 0s with NaN, then fill with median
df[invalid_zero_cols] = df[invalid_zero_cols].replace(0, np.nan)
for col in invalid_zero_cols:
    df[col] = df[col].fillna(df[col].median())

print("\n✅ Preprocessing done! Remaining NaNs:", df.isnull().sum().sum())
````

## Step 5 — Separate Features (X) & Target (y), then Train-Test Split

````python
# ================================================================
# FEATURES (X), TARGET (y), TRAIN-TEST SPLIT
# ================================================================
X = df.drop('Outcome', axis=1)
y = df['Outcome']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y             # ✅ preserve diabetes ratio
)

print(f"Training patients: {X_train.shape[0]}")
print(f"Testing patients : {X_test.shape[0]}")
print(f"Total features   : {X_train.shape[1]}")
````

# Step 6 — Feature Scaling ✅

### 💡 *Why scale here?*
Our ensemble uses **Logistic Regression** and **KNN** — both need scaling.
Tree-based models (Decision Tree, Random Forest) don't need it, but scaling doesn't hurt them either.

So we scale once and use the scaled data for everything!

````python
# ================================================================
# FEATURE SCALING
# ================================================================
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print("✅ Feature scaling done!")
print(f"Example row BEFORE: {X_train.iloc[0].values[:4].round(2)}")
print(f"Example row AFTER : {X_train_scaled[0][:4].round(2)}")
````

# Step 7 — Train Individual Base Models 🏗️

Before combining, let's see how each individual model performs alone. This gives us a **baseline** to compare against!

### Our 4 Base Models:
1. **Logistic Regression** — linear model, fast & interpretable.
2. **Decision Tree** — captures non-linear patterns.
3. **KNN** — distance-based, no assumptions about data.
4. **Random Forest** — already an ensemble (bagging of trees)!

> **Why diverse models?** The ensemble works best when models make DIFFERENT kinds of mistakes. If they all fail on the same cases, combining them doesn't help!

````python
# ================================================================
# TRAIN INDIVIDUAL BASE MODELS
# ================================================================

# Create the base models
log_reg = LogisticRegression(max_iter=1000, random_state=42)
dec_tree = DecisionTreeClassifier(max_depth=5, random_state=42)
knn = KNeighborsClassifier(n_neighbors=7)
rf = RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1)

# Store them in a dictionary
base_models = {
    'Logistic Regression': log_reg,
    'Decision Tree': dec_tree,
    'KNN': knn,
    'Random Forest': rf
}

# Train and evaluate each individually
print("=" * 65)
print("INDIVIDUAL BASE MODEL PERFORMANCE")
print("=" * 65)

individual_results = {}
for name, model in base_models.items():
    model.fit(X_train_scaled, y_train)
    y_pred = model.predict(X_test_scaled)
    y_proba = model.predict_proba(X_test_scaled)[:, 1]
    
    individual_results[name] = {
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(y_test, y_pred),
        'Recall': recall_score(y_test, y_pred),
        'F1': f1_score(y_test, y_pred),
        'ROC-AUC': roc_auc_score(y_test, y_proba)
    }
    print(f"\n{name}:")
    print(f"  Accuracy : {individual_results[name]['Accuracy']:.4f}")
    print(f"  F1 Score : {individual_results[name]['F1']:.4f}")
    print(f"  ROC-AUC  : {individual_results[name]['ROC-AUC']:.4f}")

print("\n" + "=" * 65)
````

# Step 8 — 🗳️ Ensemble Method 1: VOTING CLASSIFIER

### 💡 *How does Voting work?*
1. Each base model makes a prediction.
2. **Hard Voting:** majority class wins.
3. **Soft Voting:** average the probabilities from all models → highest wins.

**Soft Voting is usually better** because it uses probability confidence, not just yes/no votes.

````python
# ================================================================
# VOTING CLASSIFIER (Hard & Soft)
# ================================================================

# Define the list of estimators to combine
estimators = [
    ('lr', LogisticRegression(max_iter=1000, random_state=42)),
    ('dt', DecisionTreeClassifier(max_depth=5, random_state=42)),
    ('knn', KNeighborsClassifier(n_neighbors=7)),
    ('rf', RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1))
]

# --- HARD VOTING ---
hard_voting = VotingClassifier(estimators=estimators, voting='hard')
hard_voting.fit(X_train_scaled, y_train)
y_pred_hard = hard_voting.predict(X_test_scaled)

# --- SOFT VOTING ---
soft_voting = VotingClassifier(estimators=estimators, voting='soft')
soft_voting.fit(X_train_scaled, y_train)
y_pred_soft = soft_voting.predict(X_test_scaled)
y_proba_soft = soft_voting.predict_proba(X_test_scaled)[:, 1]

print("=" * 55)
print("   VOTING CLASSIFIER RESULTS")
print("=" * 55)
print(f"HARD VOTING (majority wins):")
print(f"  Accuracy : {accuracy_score(y_test, y_pred_hard):.4f}")
print(f"  F1 Score : {f1_score(y_test, y_pred_hard):.4f}")

print(f"\nSOFT VOTING (average probabilities):")
print(f"  Accuracy : {accuracy_score(y_test, y_pred_soft):.4f}")
print(f"  F1 Score : {f1_score(y_test, y_pred_soft):.4f}")
print(f"  ROC-AUC  : {roc_auc_score(y_test, y_proba_soft):.4f}")
print("=" * 55)
````

# Step 9 — 🎒 Ensemble Method 2: BAGGING CLASSIFIER

### 💡 *How does Bagging work?*
**BAGGING = Bootstrap AGGregatING**

1. Create MANY random subsets of the training data (with replacement — some rows appear multiple times!).
2. Train the SAME algorithm on each subset.
3. Combine their predictions (majority vote).

**Random Forest is actually just Bagging with Decision Trees!** But we can bag ANY algorithm.

````python
# ================================================================
# BAGGING CLASSIFIER (using Decision Tree as the base)
# ================================================================
bagging = BaggingClassifier(
    estimator=DecisionTreeClassifier(max_depth=5, random_state=42),
    n_estimators=50,          # 50 trees
    max_samples=0.8,           # each tree sees 80% of data
    bootstrap=True,            # sample WITH replacement
    random_state=42,
    n_jobs=-1
)

bagging.fit(X_train_scaled, y_train)
y_pred_bag = bagging.predict(X_test_scaled)
y_proba_bag = bagging.predict_proba(X_test_scaled)[:, 1]

print("=" * 55)
print("   BAGGING CLASSIFIER RESULTS")
print("=" * 55)
print(f"Number of estimators (trees): {bagging.n_estimators}")
print(f"\nAccuracy : {accuracy_score(y_test, y_pred_bag):.4f}")
print(f"F1 Score : {f1_score(y_test, y_pred_bag):.4f}")
print(f"ROC-AUC  : {roc_auc_score(y_test, y_proba_bag):.4f}")
print("=" * 55)

print("\n💡 Bagging reduces variance — it stabilizes unstable models like Decision Trees!")
````

# Step 10 — 🏗️ Ensemble Method 3: STACKING CLASSIFIER

### 💡 *How does Stacking work?*
Stacking is the **most powerful** ensemble technique!

1. Train several base models (Level 0).
2. Get their predictions on the training data.
3. Feed those predictions into a **meta-model** (Level 1).
4. The meta-model LEARNS how to best combine the base predictions!

**Real-life analogy:** 
- Base models = different specialist doctors (cardiologist, endocrinologist, etc.)
- Meta-model = the head physician who combines all their opinions to make the final call!

````python
# ================================================================
# STACKING CLASSIFIER
# ================================================================

# Base models (Level 0)
base_estimators = [
    ('lr', LogisticRegression(max_iter=1000, random_state=42)),
    ('dt', DecisionTreeClassifier(max_depth=5, random_state=42)),
    ('knn', KNeighborsClassifier(n_neighbors=7)),
    ('rf', RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1))
]

# Meta-model (Level 1) — combines the base predictions
meta_model = LogisticRegression(max_iter=1000, random_state=42)

stacking = StackingClassifier(
    estimators=base_estimators,
    final_estimator=meta_model,
    cv=5,               # 5-fold cross-validation to avoid overfitting
    n_jobs=-1
)

print("⏳ Training Stacking Classifier... (this trains base models + meta-model)")
stacking.fit(X_train_scaled, y_train)

y_pred_stack = stacking.predict(X_test_scaled)
y_proba_stack = stacking.predict_proba(X_test_scaled)[:, 1]

print("\n" + "=" * 55)
print("   STACKING CLASSIFIER RESULTS")
print("=" * 55)
print(f"Number of base models: {len(base_estimators)}")
print(f"Meta-model            : Logistic Regression")
print(f"\nAccuracy : {accuracy_score(y_test, y_pred_stack):.4f}")
print(f"F1 Score : {f1_score(y_test, y_pred_stack):.4f}")
print(f"ROC-AUC  : {roc_auc_score(y_test, y_proba_stack):.4f}")
print("=" * 55)
````

# Step 11 — 🏆 THE BIG COMPARISON!

The moment of truth — do the ensembles beat the individual models?

````python
# ================================================================
# FINAL COMPARISON: Individuals vs Ensembles
# ================================================================

# Collect all results
all_results = {}

# Individual models
for name, results in individual_results.items():
    all_results[name] = results

# Ensemble models
all_results['🗳️ Hard Voting'] = {
    'Accuracy': accuracy_score(y_test, y_pred_hard),
    'Precision': precision_score(y_test, y_pred_hard),
    'Recall': recall_score(y_test, y_pred_hard),
    'F1': f1_score(y_test, y_pred_hard),
    'ROC-AUC': None      # Hard voting doesn't give probabilities
}
all_results['🗳️ Soft Voting'] = {
    'Accuracy': accuracy_score(y_test, y_pred_soft),
    'Precision': precision_score(y_test, y_pred_soft),
    'Recall': recall_score(y_test, y_pred_soft),
    'F1': f1_score(y_test, y_pred_soft),
    'ROC-AUC': roc_auc_score(y_test, y_proba_soft)
}
all_results['🎒 Bagging'] = {
    'Accuracy': accuracy_score(y_test, y_pred_bag),
    'Precision': precision_score(y_test, y_pred_bag),
    'Recall': recall_score(y_test, y_pred_bag),
    'F1': f1_score(y_test, y_pred_bag),
    'ROC-AUC': roc_auc_score(y_test, y_proba_bag)
}
all_results['🏗️ Stacking'] = {
    'Accuracy': accuracy_score(y_test, y_pred_stack),
    'Precision': precision_score(y_test, y_pred_stack),
    'Recall': recall_score(y_test, y_pred_stack),
    'F1': f1_score(y_test, y_pred_stack),
    'ROC-AUC': roc_auc_score(y_test, y_proba_stack)
}

# Convert to DataFrame
results_df = pd.DataFrame(all_results).T.round(4)
print("=" * 80)
print("FINAL COMPARISON: Individual Models vs Ensembles")
print("=" * 80)
print(results_df)
````

````python
# Visualize the comparison
results_df_plot = results_df.drop(columns=['ROC-AUC']).astype(float)

plt.figure(figsize=(13, 6))
results_df_plot.plot(kind='bar', ax=plt.gca(), colormap='viridis', edgecolor='black')
plt.title('Model Performance Comparison — Individuals vs Ensembles')
plt.xlabel('Model')
plt.ylabel('Score')
plt.xticks(rotation=30, ha='right')
plt.legend(title='Metric', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.ylim(0.5, 1.0)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
````

````python
# Find the best model by F1 score
best_model_name = results_df['F1'].idxmax()
best_f1 = results_df.loc[best_model_name, 'F1']

print("=" * 55)
print(f"🏆 WINNER: {best_model_name}")
print(f"   F1 Score: {best_f1:.4f}")
print("=" * 55)

# Compare best ensemble vs best individual
individual_names = list(individual_results.keys())
ensemble_names = ['🗳️ Soft Voting', '🎒 Bagging', '🏗️ Stacking']

best_individual_f1 = results_df.loc[individual_names, 'F1'].max()
best_ensemble_f1 = results_df.loc[ensemble_names, 'F1'].max()

improvement = ((best_ensemble_f1 - best_individual_f1) / best_individual_f1) * 100

print(f"\nBest Individual F1 : {best_individual_f1:.4f}")
print(f"Best Ensemble F1   : {best_ensemble_f1:.4f}")
print(f"Improvement        : {improvement:+.2f}%")

if improvement > 0:
    print("\n🎉 The ensemble WINS! Teamwork pays off!")
else:
    print("\n🤔 The individual model won this time — ensembles don't always help!")
````

````python
# Confusion matrix for the best model
best_predictions = {
    'Logistic Regression': base_models['Logistic Regression'].predict(X_test_scaled),
    'Decision Tree': base_models['Decision Tree'].predict(X_test_scaled),
    'KNN': base_models['KNN'].predict(X_test_scaled),
    'Random Forest': base_models['Random Forest'].predict(X_test_scaled),
    '🗳️ Hard Voting': y_pred_hard,
    '🗳️ Soft Voting': y_pred_soft,
    '🎒 Bagging': y_pred_bag,
    '🏗️ Stacking': y_pred_stack
}

best_pred = best_predictions[best_model_name]
cm = confusion_matrix(y_test, best_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Greens',
            xticklabels=['Pred No Diabetes', 'Pred Diabetes'],
            yticklabels=['Actual No Diabetes', 'Actual Diabetes'])
plt.title(f"Confusion Matrix — Winner: {best_model_name}")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

print("\n📋 Detailed Classification Report:")
print(classification_report(y_test, best_pred, target_names=['No Diabetes (0)', 'Diabetes (1)']))
````

### 🏆 Golden Rules for Ensemble Learning
1. **Diverse base models = stronger ensemble** — use models with different weaknesses.
2. **Soft voting > Hard voting** (usually) — uses probability confidence.
3. **Bagging reduces variance** — great for high-variance models like Decision Trees.
4. **Boosting reduces bias** — great for weak learners (already covered in AdaBoost/GB/XGBoost).
5. **Stacking is the most powerful** — but also the most complex.
6. **Ensembles don't ALWAYS win** — sometimes a single well-tuned model is enough!
7. **More models = slower training** — balance accuracy vs speed.

### 🌟 Pros & Cons of Ensemble Learning

| ✅ Pros | ❌ Cons |
|---|---|
| **Better accuracy** than individual models | Slower to train (multiple models!) |
| **More robust** to noise & outliers | Harder to interpret ("black box") |
| Reduces overfitting (bagging) | More memory usage |
| Reduces underfitting (boosting) | More hyperparameters to tune |
| Combines strengths of multiple algorithms | Not always better than best single model |
| Won most Kaggle competitions! 🏆 | Complexity for deployment |

### 🎯 Key Concepts to Remember
1. **Ensemble Learning** = combining multiple models for better predictions.
2. **Bagging** = same algorithm, different data subsets (Random Forest is bagging!).
3. **Boosting** = sequential learning, each fixing previous errors (AdaBoost, GB, XGBoost).
4. **Voting** = different algorithms vote on the answer.
5. **Stacking** = a meta-model learns to combine base predictions.
6. **Hard vs Soft Voting** = class votes vs probability averages.
7. **Diversity matters** — models must make different mistakes!
