# 📊 Evaluation Metrics — Complete Guide
---
## Confusion Matrix
```
                Predicted
             Pos      Neg
Actual Pos [ TP  |  FN ]  ← Missed!
       Neg [ FP  |  TN ]  ← False Alarm!
```
- **TP:** Correctly said YES ✅
- **TN:** Correctly said NO ✅
- **FP:** False Alarm (Type I) ❌
- **FN:** Missed it (Type II) ❌

## Metrics
| Metric | Formula | When |
|--------|---------|------|
| Accuracy | (TP+TN)/All | Balanced data |
| Precision | TP/(TP+FP) | FP costly (spam) |
| Recall | TP/(TP+FN) | FN costly (disease) |
| F1 | 2*P*R/(P+R) | Imbalanced |
| AUC | Area under ROC | Compare models |

## Regression
| Metric | Meaning |
|--------|--------|
| MAE | Avg |error|, robust |
| MSE | Avg error², penalizes big errors |
| RMSE | √MSE, same unit |
| R² | Variance explained (1=perfect) |

````python
import numpy as np, matplotlib.pyplot as plt, seaborn as sns
from sklearn.metrics import (confusion_matrix, ConfusionMatrixDisplay,
    roc_curve, roc_auc_score, precision_recall_curve, accuracy_score,
    precision_score, recall_score, f1_score, classification_report)

# Simulated predictions
np.random.seed(42)
y_true = np.array([1]*30 + [0]*70)
y_prob = np.clip(np.random.beta(2, 5, 100) + y_true*0.3, 0, 1)
y_pred = (y_prob >= 0.5).astype(int)

print('📊 CLASSIFICATION METRICS:')
print(f'Accuracy:  {accuracy_score(y_true, y_pred):.4f}')
print(f'Precision: {precision_score(y_true, y_pred):.4f}')
print(f'Recall:    {recall_score(y_true, y_pred):.4f}')
print(f'F1 Score:  {f1_score(y_true, y_pred):.4f}')
print(f'AUC:       {roc_auc_score(y_true, y_prob):.4f}')
print(f'\n{classification_report(y_true, y_pred)}')
````

````python
# Confusion Matrix Visualization
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

cm = confusion_matrix(y_true, y_pred)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[0],
            xticklabels=['Predicted 0', 'Predicted 1'],
            yticklabels=['Actual 0', 'Actual 1'])
axes[0].set_title('Confusion Matrix', fontweight='bold')

# ROC Curve
fpr, tpr, _ = roc_curve(y_true, y_prob)
auc = roc_auc_score(y_true, y_prob)
axes[1].plot(fpr, tpr, 'b-', linewidth=2, label=f'AUC={auc:.4f}')
axes[1].plot([0,1], [0,1], 'r--', label='Random')
axes[1].set_xlabel('False Positive Rate')
axes[1].set_ylabel('True Positive Rate')
axes[1].set_title('ROC Curve', fontweight='bold')
axes[1].legend()

plt.tight_layout()
plt.show()
````

````python
# Regression Metrics Demo
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

y_true_reg = np.array([10, 20, 30, 40, 50])
y_pred_reg = np.array([12, 18, 33, 38, 52])

print('📈 REGRESSION METRICS:')
print(f'MAE:  {mean_absolute_error(y_true_reg, y_pred_reg):.2f}  (Avg absolute error)')
print(f'MSE:  {mean_squared_error(y_true_reg, y_pred_reg):.2f}  (Avg squared error)')
print(f'RMSE: {np.sqrt(mean_squared_error(y_true_reg, y_pred_reg)):.2f}  (Same unit as target)')
print(f'R2:   {r2_score(y_true_reg, y_pred_reg):.4f}  (Variance explained)')

print('\n💡 Memory Trick:')
print('  MAE  → Easy to understand')
print('  RMSE → Punishes big mistakes more')
print('  R2   → 1.0 = Perfect, 0.0 = Useless')
````
