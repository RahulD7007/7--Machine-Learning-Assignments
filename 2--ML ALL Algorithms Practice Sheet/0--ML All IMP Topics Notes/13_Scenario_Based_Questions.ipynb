# 🎭 Scenario-Based ML Interview Questions
---
These test your REAL-WORLD problem-solving skills.

## Scenario 1: Overfitting Disaster
**Q: Your model has 99% training accuracy but 60% test accuracy. What happened and how do you fix it?**

**A:**
- **Diagnosis:** Classic OVERFITTING. Model memorized training data.
- **Causes:** Model too complex, too little data, no regularization.
- **Fixes:**
  1. Get MORE training data
  2. Use a SIMPLER model (reduce max_depth, fewer features)
  3. Add REGULARIZATION (L1/L2, dropout)
  4. Use CROSS-VALIDATION
  5. Apply EARLY STOPPING
  6. Try DATA AUGMENTATION
  7. Use ENSEMBLE methods (Random Forest instead of single tree)

## Scenario 2: Imbalanced Data Crisis
**Q: Your client says the model must have >90% recall for fraud detection. How do you achieve this?**

**A:**
- **Lower threshold:** Instead of 0.5, use 0.3 (predict fraud more aggressively)
- **SMOTE:** Create synthetic fraud samples
- **Class weights:** `class_weight='balanced'` or `scale_pos_weight`
- **Undersample majority:** Reduce non-fraud samples
- **Use right metric:** Optimize for Recall, not Accuracy
- **Anomaly detection:** Isolation Forest, One-Class SVM
- **Cost-sensitive learning:** Penalize FN more than FP
- **Trade-off:** Accept lower precision (more false alarms) for higher recall

## Scenario 3: Too Many Features
**Q: You have 500 features but only 1000 samples. What do you do?**

**A:**
- **Curse of dimensionality!** More features than samples = overfitting guaranteed.
- **Solutions:**
  1. **PCA** — Reduce to top components explaining 95% variance
  2. **Lasso (L1)** — Automatically eliminates useless features
  3. **Feature selection** — SelectKBest, RFE, correlation filter
  4. **Remove correlated features** — If 2 features >0.9 correlated, drop one
  5. **Domain knowledge** — Ask business which features matter
  6. **Get more data** — If possible
  7. **Regularization** — Ridge/Lasso to prevent overfitting

## Scenario 4: Production Failure
**Q: Your model works great in testing but fails in production. Why?**

**A:**
1. **Data Leakage** — Test info leaked into training during preprocessing
2. **Data Drift** — Production data is different from training data
3. **Preprocessing Mismatch** — Scaling/encoding done differently in production
4. **Missing Features** — Production data missing columns used in training
5. **Time Leakage** — Used future data to predict past
6. **Label Mismatch** — Definition of target changed
7. **Serving Skew** — Training pipeline ≠ serving pipeline

**Prevention:**
- Use PIPELINES for consistent preprocessing
- Monitor data drift in production
- Shadow test (run new model alongside old one)
- A/B test before full rollout

## Scenario 5: Credit Card Fraud Detection
**Q: How would you build an ML system to detect credit card fraud?**

**A:**
1. **Data:** Transaction amount, time, location, merchant, user history
2. **Challenge:** Extreme imbalance (0.1% fraud!)
3. **Approach:**
   - Anomaly detection (Isolation Forest) + supervised (XGBoost)
   - SMOTE + undersampling
   - Optimize for RECALL (catch all fraud)
4. **Features:** Transaction frequency, amount deviation from user average, time patterns, location changes
5. **Real-time:** Need fast inference (<100ms) → LightGBM or logistic regression
6. **Monitoring:** Track new fraud patterns, retrain weekly
7. **Human review:** High-risk transactions → manual review queue

## Scenario 6: Deep Learning vs Traditional ML
**Q: Your manager says "just use Deep Learning for everything." What do you say?**

**A:**
- **DL needs LOTS of data** (millions+ samples). We have 10K rows.
- **For tabular/structured data**, tree-based models (XGBoost, LightGBM) usually OUTPERFORM deep learning.
- **DL shines in:** Images, text, audio, sequential data.
- **DL drawbacks:** Black box, slow training, needs GPU, hard to interpret.
- **My approach:** Start simple (Logistic Regression → Random Forest → XGBoost). Only go complex if simple models aren't good enough.
- **Quote:** "No Free Lunch Theorem — no single algorithm is best for all problems."

## Scenario 7: Explaining to Non-Technical Stakeholder
**Q: How do you explain your model to a non-technical stakeholder?**

**A:**
- **Avoid jargon.** Don't say "AUC-ROC" → say "the model correctly identifies 92% of risky loans."
- **Use analogies:** "The model is like an experienced loan officer who looks at income, credit history, and loan amount to decide risk."
- **Show top 3 factors:** "The biggest reason this loan was flagged: credit score below 600."
- **Use visuals:** Feature importance bar chart, confusion matrix as "correct vs mistakes."
- **Business impact:** "This model saves $2M/year by catching defaults early."
- **SHAP values:** "For THIS specific customer, here's why the model said 'high risk'."

## Scenario 8: Model Selection Under Constraints
**Q: You need a model that predicts in <10ms, is interpretable, and has >85% accuracy. What do you choose?**

**A:**
- **Logistic Regression** — Fastest inference, fully interpretable (coefficients), good accuracy for linear problems.
- **Small Decision Tree** — Interpretable (can show the tree), fast, but may not reach 85%.
- **NOT XGBoost/Deep Learning** — Too complex for interpretability requirement.
- **Approach:** Start with Logistic Regression. If accuracy <85%, try small Random Forest with limited trees and explain with SHAP.

## Scenario 9: Data Quality Issues
**Q: 30% of your data has missing values in a critical feature. What do you do?**

**A:**
1. **Understand WHY missing:** Is it random (MCAR) or systematic (MNAR)?
   - MCAR: Missing completely at random → impute safely
   - MNAR: Missing NOT at random → missingness IS information!
2. **If MCAR:** KNN imputer or iterative imputer (better than mean)
3. **If MNAR:** Create "is_missing" indicator column + impute
4. **If >50% missing:** Consider dropping the feature
5. **Domain knowledge:** Ask WHY it's missing. Maybe "no income" = "unemployed" = 0
6. **Multiple imputation:** Create multiple datasets with different imputations, average results

## Scenario 10: Time Series Data
**Q: You're predicting loan defaults but data has a time component. What changes?**

**A:**
1. **DON'T use random train-test split!** → Use time-based split
   - Train: Jan 2020 - Dec 2022
   - Test: Jan 2023 - Jun 2023
2. **Use TimeSeriesSplit** for cross-validation (not random KFold)
3. **Beware look-ahead bias:** Don't use future data to predict past
4. **Feature engineering:** Rolling averages, lag features, seasonality
5. **Concept drift:** Customer behavior changes over time → retrain regularly
