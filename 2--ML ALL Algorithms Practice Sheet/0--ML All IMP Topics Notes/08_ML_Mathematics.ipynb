## Gradient Descent (Like You're 10)
You're **blindfolded on a hill**, trying to reach the bottom.
- Feel the slope under your feet
- Take a step DOWNHILL
- Repeat until you reach the bottom!

```
θ_new = θ_old - α × gradient
α = Learning Rate (step size)
  Too HIGH → Jump past minimum
  Too LOW  → Takes forever
  Just RIGHT → Reaches minimum efficiently
```

## Probability Essentials
```
Bayes' Theorem: P(A|B) = P(B|A) × P(A) / P(B)
P(A)   = Prior (before evidence)
P(A|B) = Posterior (after evidence)
P(B|A) = Likelihood
```

````python
# ============================================================================
# GRADIENT DESCENT VISUALIZATION
# ============================================================================
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Cost function: J(w) = w^2 (simple parabola)
def cost(w):
    return w ** 2

def gradient(w):
    return 2 * w

# Gradient Descent with different learning rates
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
learning_rates = [0.01, 0.1, 0.99]
titles = ['Too Slow (α=0.01)', 'Just Right (α=0.1)', 'Too Fast (α=0.99)']
colors = ['blue', 'green', 'red']

w_range = np.linspace(-5, 5, 200)

for ax, lr, title, color in zip(axes, learning_rates, titles, colors):
    ax.plot(w_range, cost(w_range), 'gray', linewidth=2, label='Cost J(w)')

    w = 4.5  # Start point
    path = [w]
    for _ in range(20):
        w = w - lr * gradient(w)
        path.append(w)
        if abs(w) < 0.001:
            break

    path = np.array(path)
    ax.plot(path, cost(path), f'{color}o-', markersize=6, label=f'GD path')
    ax.set_title(title, fontweight='bold', color=color)
    ax.set_xlabel('w (parameter)')
    ax.set_ylabel('J(w) (cost)')
    ax.legend(fontsize=8)

plt.suptitle('Gradient Descent: Effect of Learning Rate',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
````

````python
# ============================================================================
# GRADIENT DESCENT FROM SCRATCH (Linear Regression)
# ============================================================================
print('📐 Gradient Descent from Scratch')
print('=' * 60)

np.random.seed(42)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)  # True: y = 4 + 3x

# GD Parameters
lr = 0.1
n_iterations = 1000
m = len(X)

# Initialize
theta = np.random.randn(2, 1)  # [intercept, slope]
X_b = np.c_[np.ones((m, 1)), X]  # Add bias term

cost_history = []
for iteration in range(n_iterations):
    gradients = 2/m * X_b.T.dot(X_b.dot(theta) - y)
    theta = theta - lr * gradients
    cost = np.mean((X_b.dot(theta) - y) ** 2)
    cost_history.append(cost)

print(f'True parameters:     intercept=4.0, slope=3.0')
print(f'GD found parameters: intercept={theta[0][0]:.4f}, slope={theta[1][0]:.4f}')
print(f'Final cost (MSE): {cost_history[-1]:.4f}')

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))
ax1.scatter(X, y, alpha=0.3, s=20)
ax1.plot(X, X_b.dot(theta), 'r-', linewidth=2, label='GD Fit')
ax1.set_title('Linear Regression via Gradient Descent', fontweight='bold')
ax1.legend()

ax2.plot(cost_history, color='blue')
ax2.set_title('Cost Decreasing Over Iterations', fontweight='bold')
ax2.set_xlabel('Iteration')
ax2.set_ylabel('MSE Cost')
plt.tight_layout()
plt.show()
````

````python
# ============================================================================
# TYPES OF GRADIENT DESCENT
# ============================================================================
print('📐 Types of Gradient Descent')
print('=' * 60)
print('''
1. BATCH GD:
   - Uses ALL data per step
   - Stable but SLOW for large data
   - Like reading entire textbook before answering

2. STOCHASTIC GD (SGD):
   - Uses ONE sample per step
   - Fast but NOISY (jumpy path)
   - Like answering after reading one page

3. MINI-BATCH GD (Best of both!):
   - Uses SMALL BATCH (32, 64, 128) per step
   - Fast AND stable
   - Like answering after reading one chapter
   - Most commonly used in Deep Learning!
''')

# Visualize the difference
fig, ax = plt.subplots(figsize=(8, 6))
w_range = np.linspace(-3, 3, 200)
ax.plot(w_range, cost(w_range), 'gray', linewidth=2)

np.random.seed(42)
for name, lr_val, color, marker in [
    ('Batch', 0.1, 'blue', 'o'),
    ('Stochastic', 0.05, 'red', 'x'),
]:
    w = 2.5
    path = [w]
    for _ in range(15):
        noise = np.random.randn() * (0.5 if 'Stoch' in name else 0)
        w = w - lr_val * (gradient(w) + noise)
        path.append(w)
    path = np.array(path)
    ax.plot(path, cost(path), f'{color}{marker}-',
            markersize=6, label=name, alpha=0.7)

ax.set_title('Batch vs Stochastic GD', fontweight='bold')
ax.legend()
plt.tight_layout()
plt.show()
````

````python
# ============================================================================
# PROBABILITY & BAYES THEOREM
# ============================================================================
print('📐 Bayes Theorem — Real Example')
print('=' * 60)

# Medical Test Example
P_disease = 0.01       # 1% have disease
P_positive_given_disease = 0.99  # Test is 99% accurate for sick
P_positive_given_healthy = 0.05  # 5% false positive rate

# P(Disease | Positive Test) = ?
P_positive = (P_positive_given_disease * P_disease +
              P_positive_given_healthy * (1 - P_disease))

P_disease_given_positive = (P_positive_given_disease * P_disease) / P_positive

print(f'P(Disease) = {P_disease} (1% of population)')
print(f'P(Positive | Disease) = {P_positive_given_disease} (99% accurate)')
print(f'P(Positive | Healthy) = {P_positive_given_healthy} (5% false positive)')
print(f'\n❓ You tested POSITIVE. What is P(Disease | Positive)?')
print(f'\n🧮 Calculation:')
print(f'   = P(+|D) × P(D) / P(+)')
print(f'   = {P_positive_given_disease} × {P_disease} / {P_positive:.4f}')
print(f'   = {P_disease_given_positive:.4f} ({P_disease_given_positive*100:.1f}%)')
print(f'\n💡 SURPRISING! Even with 99% accurate test,')
print(f'   only {P_disease_given_positive*100:.1f}% chance you actually have it!')
print(f'   Because the disease is RARE (1%), false positives dominate.')
print(f'   This is why doctors do CONFIRMATORY tests!')
````

````python
# ============================================================================
# NORMAL DISTRIBUTION & CLT
# ============================================================================
from scipy import stats

print('📐 Normal Distribution & Central Limit Theorem')
print('=' * 60)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# 1. Normal Distribution
x = np.linspace(-4, 4, 200)
for mu, sigma, color in [(0, 1, 'blue'), (0, 2, 'red'), (2, 1, 'green')]:
    axes[0].plot(x, stats.norm.pdf(x, mu, sigma), color,
                 label=f'μ={mu}, σ={sigma}')
axes[0].set_title('Normal Distributions', fontweight='bold')
axes[0].legend()

# 2. CLT Demo
np.random.seed(42)
sample_means = [np.mean(np.random.exponential(2, 30)) for _ in range(10000)]
axes[1].hist(sample_means, bins=50, color='steelblue', edgecolor='black',
             density=True, alpha=0.7)
axes[1].set_title('CLT: Sample Means → Normal!', fontweight='bold')
axes[1].set_xlabel('Sample Mean (n=30 from Exponential)')

# 3. 68-95-99.7 Rule
x = np.linspace(-4, 4, 200)
y = stats.norm.pdf(x, 0, 1)
axes[2].plot(x, y, 'black', linewidth=2)
axes[2].fill_between(x, y, where=(abs(x) <= 1), color='blue', alpha=0.3, label='68%')
axes[2].fill_between(x, y, where=(abs(x) <= 2), color='green', alpha=0.2, label='95%')
axes[2].fill_between(x, y, where=(abs(x) <= 3), color='red', alpha=0.1, label='99.7%')
axes[2].set_title('68-95-99.7 Rule', fontweight='bold')
axes[2].legend()

plt.tight_layout()
plt.show()

print('\n📖 Key Probability Rules:')
print('  68% data within ±1σ of mean')
print('  95% data within ±2σ of mean')
print('  99.7% data within ±3σ of mean')
print('\n  CLT: Sample means become Normal regardless of original distribution!')
print('  (when sample size ≥ 30)')
````

## 📝 Interview Probability Questions

**Q1: What is Bayes Theorem?**
→ Updates probability based on new evidence. P(A|B) = P(B|A)×P(A)/P(B)

**Q2: What is the Central Limit Theorem?**
→ Sample means approach Normal distribution as sample size increases (n≥30), regardless of original distribution.

**Q3: What is p-value?**
→ Probability of seeing results this extreme if null hypothesis is true. p < 0.05 → statistically significant.

**Q4: What is the difference between probability and likelihood?**
→ Probability: P(data | parameters). Likelihood: P(parameters | data).

**Q5: What is a confidence interval?**
→ Range of values that likely contains the true parameter. 95% CI means if we repeat 100 times, 95 intervals contain the true value.
