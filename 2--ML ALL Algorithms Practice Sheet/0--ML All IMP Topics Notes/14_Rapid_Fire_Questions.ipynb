## 🎯 Algorithm Selection Quick Guide

```
Is target a NUMBER?
├── YES → REGRESSION
│   ├── Linear relationship → Linear Regression
│   ├── Need regularization → Ridge(L2) / Lasso(L1)
│   └── Complex patterns → RF / XGBoost Regressor
│
└── NO → CLASSIFICATION
    ├── Need interpretable → Logistic Regression / Decision Tree
    ├── Small data → Naive Bayes
    ├── Best accuracy → XGBoost / LightGBM
    ├── High dimensions → SVM
    └── Baseline → Random Forest

No labels?
├── Group items → K-Means / DBSCAN
├── Reduce features → PCA
└── Find anomalies → Isolation Forest / DBSCAN
```

## 📊 Metric Selection Quick Guide

```
Balanced data → Accuracy is fine
Imbalanced data → F1, Precision, Recall, AUC
FP costly (spam) → Optimize PRECISION
FN costly (cancer) → Optimize RECALL
Compare models → ROC-AUC
Regression → RMSE, MAE, R²
```
