# 🔍 Complete EDA (Exploratory Data Analysis)
---
EDA = Being a detective with data! Find patterns, outliers, relationships BEFORE modeling.

```python
import pandas as pd, numpy as np, warnings
import matplotlib.pyplot as plt, seaborn as sns
warnings.filterwarnings('ignore')

df = pd.read_csv('loan_default_dataset.csv')
print(f'Shape: {df.shape}')
print(f'\nData Types:\n{df.dtypes}')
print(f'\nMissing:\n{df.isnull().sum()[df.isnull().sum()>0]}')
df.head()
```

```python
# Target Distribution
plt.figure(figsize=(8, 5))
df['Default'].value_counts().plot(kind='bar', color=['#2ecc71','#e74c3c'], edgecolor='black')
plt.title('Target Distribution', fontweight='bold')
plt.xticks(rotation=0)
plt.show()
print(f'Imbalance ratio: {df["Default"].value_counts().max()/df["Default"].value_counts().min():.2f}:1')
```

```python
# Numerical Distributions
num_cols = [c for c in df.select_dtypes(include=[np.number]).columns if c != 'Default']
ncols = 3
nrows = (len(num_cols)+ncols-1)//ncols
fig, axes = plt.subplots(nrows, ncols, figsize=(15, 4*nrows))
axes = axes.flatten()
for i, c in enumerate(num_cols):
    axes[i].hist(df[c].dropna(), bins=30, color='steelblue', edgecolor='black', alpha=0.7)
    axes[i].set_title(c, fontweight='bold')
for i in range(len(num_cols), len(axes)):
    axes[i].set_visible(False)
plt.suptitle('Numerical Distributions', fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Boxplots by Target
fig, axes = plt.subplots(nrows, ncols, figsize=(15, 4*nrows))
axes = axes.flatten()
for i, c in enumerate(num_cols):
    df.boxplot(column=c, by='Default', ax=axes[i])
    axes[i].set_title(f'{c} by Default')
for i in range(len(num_cols), len(axes)):
    axes[i].set_visible(False)
plt.suptitle('Boxplots by Default Status', fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
# Correlation Heatmap
corr = df.select_dtypes(include=[np.number]).corr()
plt.figure(figsize=(12, 8))
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='RdBu_r', center=0)
plt.title('Correlation Matrix', fontweight='bold')
plt.tight_layout()
plt.show()

print('Correlation with Default:')
print(corr['Default'].drop('Default').sort_values(key=abs, ascending=False).round(3))
```

```python
# Categorical Analysis
cat_cols = df.select_dtypes(include=['object']).columns.tolist()
for c in cat_cols:
    print(f'\n{c}: {df[c].nunique()} unique')
    print(df[c].value_counts())
    plt.figure(figsize=(10, 4))
    rate = df.groupby(c)['Default'].mean().sort_values(ascending=False)
    rate.plot(kind='bar', color='coral')
    plt.title(f'Default Rate by {c}', fontweight='bold')
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.show()
```
