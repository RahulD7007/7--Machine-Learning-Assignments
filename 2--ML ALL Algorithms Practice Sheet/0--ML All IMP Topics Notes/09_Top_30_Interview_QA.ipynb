# 🎯 Top 30 ML Interview Questions & Answers
---
Simple, concise answers you can use directly in interviews!

### Q1: What is the difference between Supervised and Unsupervised Learning?
**A:** Supervised = We have labels (answers). Model learns INPUT→OUTPUT mapping. Like studying with answer key.
Unsupervised = No labels. Model finds hidden PATTERNS/GROUPS on its own. Like organizing a messy room without instructions.

### Q2: What is overfitting and how do you prevent it?
**A:** Overfitting = Model memorizes training data but fails on new data. Like memorizing answers without understanding.
**Prevention:** More data, cross-validation, regularization (L1/L2), pruning, dropout, early stopping, simpler model.

### Q3: What is the difference between Classification and Regression?
**A:** Classification → Predicts CATEGORIES (spam/not spam, cat/dog). Regression → Predicts NUMBERS (price, temperature).

### Q4: Explain Bias-Variance Tradeoff.
**A:** Bias = error from wrong assumptions (too simple). Variance = error from sensitivity to training data (too complex). We need BALANCE — not too simple, not too complex.

### Q5: What is regularization? Why do we need it?
**A:** Adds PENALTY for complexity. Prevents overfitting. L1 (Lasso) → eliminates features. L2 (Ridge) → shrinks all coefficients.

### Q6: When would you use Random Forest vs XGBoost?
**A:** RF: Good baseline, less overfitting, easier to tune. XGBoost: Usually higher accuracy, needs more tuning, risk of overfitting. Start with RF, then try XGBoost.

### Q7: Explain Precision vs Recall with example.
**A:** Precision = "Of all my YES predictions, how many were correct?" Recall = "Of all actual YES cases, how many did I find?" Cancer → Recall more important. Spam → Precision more important.

### Q8: What is the curse of dimensionality?
**A:** As features increase, data becomes SPARSE. Distance measures become meaningless. Need exponentially more data. Solution: PCA, feature selection.

### Q9: How do you handle missing data?
**A:** Remove (if few), impute (mean/median/mode), KNN imputer, create missing indicator. Depends on amount and mechanism (MCAR, MAR, MNAR).

### Q10: What is cross-validation and why important?
**A:** Split data into K parts, rotate test fold, average scores. More reliable than single split. Use Stratified K-Fold for imbalanced data.

### Q11: Explain the ROC-AUC curve.
**A:** ROC plots TPR vs FPR at different thresholds. AUC = area under curve. 1.0=perfect, 0.5=random. Useful for comparing models with imbalanced data.

### Q12: What is feature scaling and when needed?
**A:** Making features comparable in range. Needed for: distance-based (KNN, SVM), gradient-based (LR, NN). NOT needed for tree-based models.

### Q13: What is multicollinearity?
**A:** Two+ features highly correlated. Problem: unstable coefficients. Detection: VIF > 5. Solution: remove one, PCA, Ridge.

### Q14: Explain ensemble learning methods.
**A:** Bagging (parallel, reduces variance) → RF. Boosting (sequential, reduces bias) → XGBoost. Stacking (combine different models → meta-learner).

### Q15: What is data leakage?
**A:** Training data accidentally contains test/future info. Model performs great in testing, terrible in production. Prevention: split first, then preprocess, use pipelines.

### Q16: How do you handle imbalanced datasets?
**A:** SMOTE, undersampling, class weights, different threshold, use F1/AUC instead of accuracy, stratified K-Fold.

### Q17: Parametric vs non-parametric models?
**A:** Parametric: fixed form, fixed params (LR, NB). Non-parametric: no assumptions, params grow with data (KNN, DT, SVM).

### Q18: What is gradient descent?
**A:** Optimization to find parameters minimizing error. Like walking downhill blindfolded. Learning rate = step size.

### Q19: Difference between bagging and boosting?
**A:** Bagging: independent trees, parallel, reduces variance. Boosting: sequential trees, each corrects previous, reduces bias.

### Q20: How would you choose between models?
**A:** Start simple (LR baseline). Try tree-based (RF, XGBoost). Compare on validation data. Consider: interpretability, speed, data size.

### Q21: L1 vs L2 regularization?
**A:** L1 (Lasso): absolute values, creates sparse models (feature selection). L2 (Ridge): squared values, shrinks all but never zero. ElasticNet: both.

### Q22: Explain Kernel Trick in SVM.
**A:** Maps data to higher dimension where linearly separable, WITHOUT computing transformation directly. RBF kernel most common.

### Q23: Hyperparameters vs Parameters?
**A:** Parameters: learned BY model (weights). Hyperparameters: set BY us before training (learning rate, n_trees, max_depth).

### Q24: What is one-hot encoding? When to use?
**A:** Convert categories to binary columns. Use for nominal data (no order). Avoid for high-cardinality (too many columns).

### Q25: What is the Elbow Method?
**A:** Plot K vs inertia for K-Means. Choose K at the "elbow" where adding clusters gives diminishing returns.

### Q26: Explain R-squared.
**A:** Proportion of variance explained by model. R²=1 perfect, R²=0 no better than mean. Can be negative (worse than mean!).

### Q27: StandardScaler vs MinMaxScaler?
**A:** Standard: mean=0, std=1, robust to outliers. MinMax: [0,1], sensitive to outliers.

### Q28: What is early stopping?
**A:** Stop training when validation performance stops improving. Prevents overfitting. Used in boosting and neural networks.

### Q29: What is a pipeline in ML?
**A:** Chains preprocessing + modeling into one object. Prevents data leakage, ensures consistent preprocessing, easier to deploy.

### Q30: How do you deploy an ML model?
**A:** Train → Save (pickle/joblib) → API (Flask/FastAPI) → Docker → AWS/Heroku → Monitor for drift.
