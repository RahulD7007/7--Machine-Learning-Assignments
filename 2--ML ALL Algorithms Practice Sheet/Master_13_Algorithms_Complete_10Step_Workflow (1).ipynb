```python
# =====================================================================
# MASTER IMPORT: ALL COMMON LIBRARIES
# =====================================================================
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Scikit-Learn Common Tools
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, MinMaxScaler

sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (7, 4)
print("✅ Master Setup Ready: All core libraries loaded!")
```

# =========================================================================
# 🍕 ALGORITHM 1: LINEAR REGRESSION
# =========================================================================

### 🍕 8th-Grade Story: The Pizza Price Equation
A plain pizza base costs **$5.00** (Intercept $b$), and each extra topping adds **$2.00** (Slope $w$).
$$\mathbf{y = w_1 x_1 + w_2 x_2 + \dots + b}$$
Linear Regression draws the **best straight ruler line** through continuous data points!

```python
# --- 1. Import Specific Algorithm ---
from sklearn.datasets import fetch_california_housing
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# --- 2. Load Dataset ---
california = fetch_california_housing(as_frame=True)
df1 = california.frame.head(500).copy() # Using 500 samples for fast demo
X1 = df1[['MedInc', 'AveRooms', 'HouseAge', 'Population']]
y1 = df1['MedHouseVal'] # Price in $100,000s

# --- 3. Exploratory Data Analysis (EDA) ---
print("1. Linear Regression Data Shape:", X1.shape)
print("Correlation with Price:\n", df1[['MedInc', 'AveRooms', 'MedHouseVal']].corr()['MedHouseVal'])

# --- 4. Data Preprocessing ---
X1 = X1.fillna(X1.median())

# --- 5. Train-Test Split (80/20) ---
X1_train, X1_test, y1_train, y1_test = train_test_split(X1, y1, test_size=0.20, random_state=42)

# --- 6. Feature Scaling ---
scaler1 = StandardScaler()
X1_train_scaled = scaler1.fit_transform(X1_train)
X1_test_scaled = scaler1.transform(X1_test)

# --- 7. Model Selection & Training ---
model1 = LinearRegression()
model1.fit(X1_train_scaled, y1_train)

# --- 8. Prediction ---
y1_pred = model1.predict(X1_test_scaled)

# --- 9. Model Evaluation ---
mae1 = mean_absolute_error(y1_test, y1_pred)
r2_1 = r2_score(y1_test, y1_pred)
print(f"Evaluation -> MAE: ${mae1*100000:,.0f} | R² Score: {r2_1:.3f}")

# --- 10. Hyperparameter Tuning (Ridge Regularization) ---
ridge_cv = GridSearchCV(Ridge(), {'alpha': [0.1, 1.0, 10.0, 100.0]}, cv=3)
ridge_cv.fit(X1_train_scaled, y1_train)
print(f"Tuned Ridge Best Alpha: {ridge_cv.best_params_['alpha']} | Tuned R²: {ridge_cv.best_score_:.3f}")
```

# =========================================================================
# 🚦 ALGORITHM 2: LOGISTIC REGRESSION
# =========================================================================

### 🚦 8th-Grade Story: The Rain Probability S-Curve
Unlike Linear Regression which outputs any number, Logistic Regression passes numbers through the **Sigmoid function** to output a **probability between 0% and 100%** for Yes/No choices (e.g. Will it rain? Is this tumor benign?).
$$\mathbf{P(y=1) = \frac{1}{1 + e^{-z}}}$$

```python
# --- 1. Import Specific Algorithm ---
from sklearn.datasets import load_breast_cancer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# --- 2. Load Dataset ---
cancer = load_breast_cancer(as_frame=True)
X2 = cancer.frame.drop(columns=['target'])
y2 = cancer.frame['target'] # 0 = Malignant, 1 = Benign

# --- 3. Exploratory Data Analysis (EDA) ---
print("2. Logistic Regression Target Breakdown:\n", y2.value_counts())

# --- 4. Data Preprocessing ---
X2 = X2.fillna(X2.median())

# --- 5. Train-Test Split (80/20) ---
X2_train, X2_test, y2_train, y2_test = train_test_split(X2, y2, test_size=0.20, random_state=42, stratify=y2)

# --- 6. Feature Scaling ---
scaler2 = StandardScaler()
X2_train_scaled = scaler2.fit_transform(X2_train)
X2_test_scaled = scaler2.transform(X2_test)

# --- 7. Model Selection & Training ---
model2 = LogisticRegression(max_iter=1000, random_state=42)
model2.fit(X2_train_scaled, y2_train)

# --- 8. Prediction ---
y2_pred = model2.predict(X2_test_scaled)

# --- 9. Model Evaluation ---
print(f"Evaluation -> Accuracy: {accuracy_score(y2_test, y2_pred)*100:.2f}% | F1-Score: {f1_score(y2_test, y2_pred)*100:.2f}%")

# --- 10. Hyperparameter Tuning ---
grid2 = GridSearchCV(LogisticRegression(max_iter=1000, random_state=42), {'C': [0.01, 0.1, 1.0, 10.0]}, cv=3)
grid2.fit(X2_train_scaled, y2_train)
print(f"Tuned Logistic Regression Best C: {grid2.best_params_['C']} | CV Accuracy: {grid2.best_score_*100:.2f}%")
```

# =========================================================================
# 🌳 ALGORITHM 3: DECISION TREE
# =========================================================================

### 🌳 8th-Grade Story: The 20-Questions Flowchart Game
A Decision Tree asks simple Yes/No questions: *"Is weight > 50kg? Does it have wheels?"* 
It splits data down the tree until reaching a pure answer (Gini Impurity = 0). It requires **NO feature scaling**!

```python
# --- 1. Import Specific Algorithm ---
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier, plot_tree

# --- 2. Load Dataset ---
iris = load_iris(as_frame=True)
X3 = iris.frame.drop(columns=['target'])
y3 = iris.frame['target']

# --- 3. EDA & 4. Preprocessing ---
print("3. Decision Tree Classes:", iris.target_names)

# --- 5. Train-Test Split ---
X3_train, X3_test, y3_train, y3_test = train_test_split(X3, y3, test_size=0.20, random_state=42, stratify=y3)

# --- 6. Feature Scaling (Not required for Trees, skipped!) ---

# --- 7. Model Training ---
model3 = DecisionTreeClassifier(max_depth=3, random_state=42)
model3.fit(X3_train, y3_train)

# --- 8. Prediction & 9. Evaluation ---
y3_pred = model3.predict(X3_test)
print(f"Decision Tree Accuracy: {accuracy_score(y3_test, y3_pred)*100:.2f}%")

# Visual Flowchart Tree
plt.figure(figsize=(8, 4))
plot_tree(model3, feature_names=iris.feature_names, class_names=iris.target_names, filled=True, rounded=True)
plt.title('Decision Tree Flowchart Breakdown 🌳')
plt.show()

# --- 10. Hyperparameter Tuning ---
grid3 = GridSearchCV(DecisionTreeClassifier(random_state=42), {'max_depth': [2, 3, 4, 5], 'min_samples_split': [2, 5]}, cv=3)
grid3.fit(X3_train, y3_train)
print(f"Tuned Tree Best Depth: {grid3.best_params_['max_depth']} | Score: {grid3.best_score_*100:.2f}%")
```

# =========================================================================
# 🌲 ALGORITHM 4: RANDOM FOREST
# =========================================================================

### 🌲 8th-Grade Story: 100 Students Voting in the Cafeteria
Instead of trusting 1 single decision tree (which might be biased), Random Forest creates **100 diverse trees** on random subsets and takes a **democratic majority vote** (Wisdom of Crowds)!

```python
# --- 1. Import Specific Algorithm ---
from sklearn.datasets import load_wine
from sklearn.ensemble import RandomForestClassifier

# --- 2. Load Dataset ---
wine = load_wine(as_frame=True)
X4 = wine.frame.drop(columns=['target'])
y4 = wine.frame['target']

# --- 3. EDA & 4. Preprocessing & 5. Split ---
X4_train, X4_test, y4_train, y4_test = train_test_split(X4, y4, test_size=0.20, random_state=42, stratify=y4)

# --- 6. Scaling (Skipped for Forest) & 7. Training ---
model4 = RandomForestClassifier(n_estimators=100, random_state=42)
model4.fit(X4_train, y4_train)

# --- 8. Prediction & 9. Evaluation ---
y4_pred = model4.predict(X4_test)
print(f"Random Forest (100 Trees) Accuracy: {accuracy_score(y4_test, y4_pred)*100:.2f}%")

# --- 10. Hyperparameter Tuning ---
grid4 = GridSearchCV(RandomForestClassifier(random_state=42), {'n_estimators': [50, 100], 'max_depth': [3, 5, None]}, cv=3)
grid4.fit(X4_train, y4_train)
print(f"Tuned Random Forest Best Params: {grid4.best_params_}")
```

# =========================================================================
# 🛣️ ALGORITHM 5: SUPPORT VECTOR MACHINE (SVM)
# =========================================================================

### 🛣️ 8th-Grade Story: The Dodgeball Safety Highway
SVM finds the **widest possible highway (maximum margin)** to separate two teams. The closest players touching the highway guardrails are called **Support Vectors**.
The **Kernel Trick** lifts flat non-separable 2D points into 3D space!

```python
# --- 1. Import Specific Algorithm ---
from sklearn.svm import SVC

# --- 2. Load, 3. EDA, 4. Preprocess, 5. Split ---
X5_train, X5_test, y5_train, y5_test = train_test_split(X2, y2, test_size=0.20, random_state=42, stratify=y2)

# --- 6. Feature Scaling (Crucial for SVM!) ---
scaler5 = StandardScaler()
X5_train_scaled = scaler5.fit_transform(X5_train)
X5_test_scaled = scaler5.transform(X5_test)

# --- 7. Training & 8. Prediction & 9. Evaluation ---
model5 = SVC(kernel='rbf', C=1.0, random_state=42)
model5.fit(X5_train_scaled, y5_train)
y5_pred = model5.predict(X5_test_scaled)
print(f"SVM Accuracy: {accuracy_score(y5_test, y5_pred)*100:.2f}%")

# --- 10. Hyperparameter Tuning ---
grid5 = GridSearchCV(SVC(random_state=42), {'C': [0.1, 1, 10], 'kernel': ['linear', 'rbf']}, cv=3)
grid5.fit(X5_train_scaled, y5_train)
print(f"Tuned SVM Best Settings: {grid5.best_params_} | Accuracy: {grid5.best_score_*100:.2f}%")
```

# =========================================================================
# 👥 ALGORITHM 6: K-NEAREST NEIGHBORS (KNN)
# =========================================================================

### 👥 8th-Grade Story: The 5 Closest Lunch Friends
$"\text{Tell me who your 5 closest friends at the lunch table are, and I'll tell you your hobby!}"$
KNN measures physical Euclidean distance to the nearest $K$ data points.

```python
# --- 1. Import Specific Algorithm ---
from sklearn.neighbors import KNeighborsClassifier

# --- 2. Load, 5. Split, 6. Scaling (Crucial for Distance!) ---
X6_train, X6_test, y6_train, y6_test = train_test_split(X3, y3, test_size=0.20, random_state=42, stratify=y3)
scaler6 = StandardScaler()
X6_train_scaled = scaler6.fit_transform(X6_train)
X6_test_scaled = scaler6.transform(X6_test)

# --- 7. Train & 8. Predict & 9. Evaluate ---
model6 = KNeighborsClassifier(n_neighbors=5)
model6.fit(X6_train_scaled, y6_train)
y6_pred = model6.predict(X6_test_scaled)
print(f"KNN (K=5) Accuracy: {accuracy_score(y6_test, y6_pred)*100:.2f}%")

# --- 10. Hyperparameter Tuning ---
grid6 = GridSearchCV(KNeighborsClassifier(), {'n_neighbors': [1, 3, 5, 7, 9]}, cv=3)
grid6.fit(X6_train_scaled, y6_train)
print(f"Tuned Optimal K: {grid6.best_params_['n_neighbors']} | Accuracy: {grid6.best_score_*100:.2f}%")
```

# =========================================================================
# 🦸 ALGORITHM 8: ENSEMBLE LEARNING (Voting & Stacking)
# =========================================================================

### 🦸 8th-Grade Story: The Avengers Super-Team
Combining Iron Man (Random Forest), Captain America (Logistic Regression), and Thor (SVM).
- **VotingClassifier**: Democratic majority vote.
- **StackingClassifier**: Nick Fury (Meta-Learner) learns which hero to trust in each specific battle!

```python
# --- 1. Import Specific Algorithm ---
from sklearn.ensemble import VotingClassifier, StackingClassifier

# --- 7. Assemble the Superhero Models ---
heroes = [
    ('lr', LogisticRegression(max_iter=1000, random_state=42)),
    ('rf', RandomForestClassifier(n_estimators=50, random_state=42)),
    ('svc', SVC(probability=True, random_state=42))
]

# Voting Ensemble
voting_model = VotingClassifier(estimators=heroes, voting='soft')
voting_model.fit(X5_train_scaled, y5_train)

# Stacking Ensemble
stacking_model = StackingClassifier(estimators=heroes, final_estimator=LogisticRegression())
stacking_model.fit(X5_train_scaled, y5_train)

# --- 8. Predict & 9. Evaluate ---
print(f"🦸 Soft Voting Ensemble Accuracy  : {accuracy_score(y5_test, voting_model.predict(X5_test_scaled))*100:.2f}%")
print(f"🛡️ Stacking Meta-Learner Accuracy : {accuracy_score(y5_test, stacking_model.predict(X5_test_scaled))*100:.2f}%")
```

# ⚽ ALGORITHM 12: K-MEANS CLUSTERING (UNSUPERVISED)
# =========================================================================

### ⚽ 8th-Grade Story: Gym Captains on a Soccer Field
Teacher places **$K=3$ Team Captains (Centroids)**. Students run to their closest captain. Captains move to the physical center of their team. Repeat until balanced teams form!
Evaluated using **Inertia (Elbow Method)** and **Silhouette Score** ($> 0.5$ is great).

```python
# --- 1. Import Specific Algorithm ---
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# --- 2. Load Unlabelled Dataset & 6. Scale ---
X12, _ = make_blobs(n_samples=300, centers=3, cluster_std=0.7, random_state=42)
X12_scaled = StandardScaler().fit_transform(X12)

# --- 7. Train K-Means (K=3) & 8. Predict Clusters ---
kmeans = KMeans(n_clusters=3, n_init=10, random_state=42)
labels12 = kmeans.fit_predict(X12_scaled)

# --- 9. Model Evaluation (Silhouette Score) ---
sil12 = silhouette_score(X12_scaled, labels12)
print(f"K-Means (K=3) Silhouette Score: {sil12:.3f} (Values > 0.5 mean crisp clusters! ⭐)")

# --- 10. Hyperparameter Tuning (Elbow Curve to Find Optimal K) ---
inertias = [KMeans(n_clusters=k, n_init=5, random_state=42).fit(X12_scaled).inertia_ for k in range(1, 7)]
plt.figure(figsize=(7, 3.5))
plt.plot(range(1, 7), inertias, 'bo-')
plt.title('Elbow Method: Spotting Optimal K at the Bend (K=3) ⚽')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.show()
```

# =========================================================================
# 🌌 ALGORITHM 13: DBSCAN CLUSTERING (UNSUPERVISED)
# =========================================================================

### 🌌 8th-Grade Story: Beach Parties vs Lonely Swimmers
DBSCAN connects dense points into arbitrarily shaped islands (like crescent moons 🌙). Stray isolated points are automatically flagged as **Noise / Outliers (Label = -1)**!

```python
# --- 1. Import Specific Algorithm ---
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN

# --- 2. Load Curved Dataset & 6. Scale ---
X13, _ = make_moons(n_samples=250, noise=0.08, random_state=42)
X13_scaled = StandardScaler().fit_transform(X13)

# --- 7. Train DBSCAN & 8. Predict ---
dbscan = DBSCAN(eps=0.25, min_samples=5)
db_labels = dbscan.fit_predict(X13_scaled)

# --- 9. Model Evaluation ---
n_clusters = len(set(db_labels)) - (1 if -1 in db_labels else 0)
n_noise = list(db_labels).count(-1)
print(f"DBSCAN Discovered: {n_clusters} Curved Clusters | Outlier Noise Points: {n_noise}")

# Visualizing DBSCAN Crescent Moons
plt.figure(figsize=(7, 3.5))
plt.scatter(X13_scaled[:, 0], X13_scaled[:, 1], c=db_labels, cmap='coolwarm', s=35)
plt.title('DBSCAN: Perfectly Separating Curved Crescent Moons 🌙')
plt.show()

print("\n🎉 CONGRATULATIONS! You have mastered all 13 Machine Learning Algorithms using the 10-Step Workflow!")
```
