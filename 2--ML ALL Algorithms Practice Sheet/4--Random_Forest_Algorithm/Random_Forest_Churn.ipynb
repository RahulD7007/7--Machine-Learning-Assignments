### What is Random Forest?
A Random Forest is exactly what it sounds like — a **FOREST of many Decision Trees**!
Instead of trusting one tree, it asks **hundreds of trees** and takes a **majority vote**.

## What is it?

**Real-life analogy:** Imagine deciding which movie to watch.
- Asking ONE friend = Decision Tree (could be biased!)
- Asking 100 friends and going with the majority = Random Forest (much wiser!) 🎬

### Main idea
1. Build **many** Decision Trees (e.g., 100 trees).
2. Each tree sees a **random subset** of the data (called **bootstrapping**).
3. Each tree considers only a **random subset of features** at each split.
4. For prediction: **majority vote** (classification) or **average** (regression).

This technique is called **Ensemble Learning** — combining many weak models into one strong model.

---

## 🔑 Decision Tree vs Random Forest — Key Differences

| Feature | Decision Tree 🌳 | Random Forest 🌲🌲🌲 |
|---|---|---|
| **Number of trees** | 1 | Many (e.g., 100) |
| **Accuracy** | Good | **Better** |
| **Overfitting** | High risk | Much lower (voting reduces errors) |
| **Speed** | Very fast | Slower (many trees to build) |
| **Interpretability** | Easy (visualize the tree) | Harder (100 trees to look at!) |
| **Feature Importance** | ✅ Yes | ✅ Yes (more reliable) |

# Step 4 — Data Preprocessing

### Tasks to do:
1. **Drop `customerID`** — just an ID, no predictive value.
2. **Fix `TotalCharges`** — this column looks numeric but is stored as text with hidden spaces!
3. **Encode target `Churn`** → Yes=1, No=0.
4. **One-Hot Encode** all remaining categorical (text) columns.

```python
# ----------------------------------------------------------------
# 4.1 Drop useless column
# ----------------------------------------------------------------
df = df.drop(columns=['customerID'])
print("✅ Dropped customerID column")
```

```python
# ----------------------------------------------------------------
# 4.2 Fix TotalCharges — it's stored as TEXT with some blank spaces!
# ----------------------------------------------------------------
print("Data type of TotalCharges BEFORE fixing:", df['TotalCharges'].dtype)

# Convert to numeric — 'errors=coerce' turns invalid (blank) values into NaN
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')

print("Data type AFTER conversion    :", df['TotalCharges'].dtype)
print("Number of missing TotalCharges:", df['TotalCharges'].isnull().sum())

# Fill the few missing values with the median
df['TotalCharges'] = df['TotalCharges'].fillna(df['TotalCharges'].median())

print("✅ TotalCharges fixed! Missing values now:", df['TotalCharges'].isnull().sum())
```

```python
# ----------------------------------------------------------------
# 4.3 Encode target Churn: Yes=1, No=0
# ----------------------------------------------------------------
df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})
print("✅ Churn column encoded (Yes=1, No=0)")
print(df['Churn'].value_counts())
```

```python
# ----------------------------------------------------------------
# 4.4 One-Hot Encode all categorical (text) columns
# ----------------------------------------------------------------
# Find all remaining text columns
categorical_cols = df.select_dtypes(include='object').columns.tolist()
print("Categorical columns to encode:", categorical_cols)

# One-Hot Encoding: turns each category into its own 0/1 column
df = pd.get_dummies(df, columns=categorical_cols, drop_first=True)

# Convert bool columns (from get_dummies) to int for cleanness
bool_cols = df.select_dtypes(include='bool').columns
df[bool_cols] = df[bool_cols].astype(int)

print(f"\n✅ Encoding complete! Dataset now has {df.shape[1]} columns.")
df.head()
```

```python
# Final check: no missing values, all numeric
print("Any missing values left?", df.isnull().sum().sum())
print("\nAll data types are now numeric:")
print(df.dtypes.value_counts())
```

## Step 5 — Separate Features (X) & Target (y), then Train-Test Split

- `X` = all the **clues** (input columns).
- `y` = the **answer** (`Churn`) we want to predict.

### ⚠️ IMPORTANT for Classification: Use `stratify=y`
Especially important here because our data is **imbalanced** (73% No, 27% Yes)!

```python
# ================================================================
# FEATURES (X), TARGET (y), TRAIN-TEST SPLIT
# ================================================================
X = df.drop('Churn', axis=1)   # everything EXCEPT Churn
y = df['Churn']                 # what we want to predict

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,       # 20% for testing
    random_state=42,       # reproducible split
    stratify=y             # ✅ preserve churn ratio in both sets
)

print(f"Training customers: {X_train.shape[0]}")
print(f"Testing customers : {X_test.shape[0]}")
print(f"Total features    : {X_train.shape[1]}")
print("\nTrain class balance:")
print(y_train.value_counts(normalize=True).round(2))
print("\nTest class balance:")
print(y_test.value_counts(normalize=True).round(2))
```

# Step 6 — Feature Scaling ❌ NOT NEEDED!

### 💡 *Why don't we scale for Random Forest?*
Just like a single Decision Tree, Random Forest builds **many trees**, and each tree makes decisions like:
> "Is `tenure > 12`?" or "Is `MonthlyCharges > 70`?"

The **scale doesn't matter** for split-based rules — the *ordering* of values is what counts.

> **Save yourself time — skip scaling for tree-based models!**

# Step 7 — Model Training

### 💡 *How Random Forest learns:*
1. **Bootstrap Sampling** — creates many random subsets of the training data (with replacement).
2. Build one Decision Tree per subset.
3. At each split, each tree considers only a **random subset of features** (usually `sqrt(total_features)`).
4. All trees vote → majority wins for classification.

This randomness makes the forest **diverse**, which reduces overfitting!

### Key parameters:
- `n_estimators=100` → number of trees (more = better but slower)
- `random_state=42` → reproducibility
- `n_jobs=-1` → use ALL CPU cores (much faster training!) 🚀

```python
# ================================================================
# MODEL TRAINING
# ================================================================
model = RandomForestClassifier(
    n_estimators=100,      # 100 trees in our forest
    random_state=42,        # reproducible
    n_jobs=-1               # use all CPU cores in parallel
)

# .fit() trains all 100 trees
model.fit(X_train, y_train)

print("✅ Random Forest with 100 trees trained!")
print(f"Number of trees: {len(model.estimators_)}")
print(f"Total features : {model.n_features_in_}")
```

# Step 8 — Prediction

Two useful outputs:
- `.predict()` → final class (0 or 1) via majority vote across all trees.
- `.predict_proba()` → probability of churning (fraction of trees that voted 'Yes').

```python
# ================================================================
# PREDICTION
# ================================================================
y_pred = model.predict(X_test)                     # 0 or 1
y_proba = model.predict_proba(X_test)[:, 1]        # probability of churning

# Compare actual vs predicted for first 10 test customers
comparison = pd.DataFrame({
    'Actual': y_test.values[:10],
    'Predicted': y_pred[:10],
    'Churn Probability': y_proba[:10].round(3)
})
comparison['Correct?'] = comparison['Actual'] == comparison['Predicted']
comparison
```

# Step 9 — Evaluate the Model

| Metric | Meaning |
|---|---|
| **Accuracy** | Overall correctness |
| **Precision** | Of predicted churners, how many actually churned? |
| **Recall** | Of actual churners, how many did we catch? |
| **F1 Score** | Balance of Precision & Recall |
| **ROC-AUC** | Overall separator quality (0.5 = random, 1.0 = perfect) |

### ⚠️ Business insight: In churn prediction, high **Recall** is crucial!
Missing a churner (FN) means losing a customer forever. False alarms (FP) just mean an unnecessary retention offer — much cheaper!

```python
# ================================================================
# MODEL EVALUATION
# ================================================================
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
auc = roc_auc_score(y_test, y_proba)

print("=" * 50)
print("   BASELINE RANDOM FOREST — REPORT CARD")
print("=" * 50)
print(f"  Accuracy  : {acc:.4f} ({acc*100:.1f}%)")
print(f"  Precision : {prec:.4f}")
print(f"  Recall    : {rec:.4f}")
print(f"  F1 Score  : {f1:.4f}")
print(f"  ROC-AUC   : {auc:.4f}")
print("=" * 50)

print("\n📋 Detailed Classification Report:")
print(classification_report(y_test, y_pred, target_names=['Stayed (0)', 'Churned (1)']))
```

```python
# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Pred Stayed', 'Pred Churned'],
            yticklabels=['Actual Stayed', 'Actual Churned'])
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

tn, fp, fn, tp = cm.ravel()
print(f"✅ True Negatives  (TN): {tn}  -> Correctly predicted 'Stayed'")
print(f"❌ False Positives (FP): {fp}  -> Predicted 'Churn' but stayed (false alarm)")
print(f"❌ False Negatives (FN): {fn}  -> Predicted 'Stayed' but LEFT (missed churner!)")
print(f"✅ True Positives  (TP): {tp}  -> Correctly predicted 'Churned'")
```

```python
# ROC Curve
fpr, tpr, thresholds = roc_curve(y_test, y_proba)

plt.figure(figsize=(7, 6))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'Random Forest (AUC = {auc:.3f})')
plt.plot([0, 1], [0, 1], color='gray', lw=1, linestyle='--', label='Random guess (AUC = 0.5)')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve — How well does our forest separate churners?')
plt.legend(loc='lower right')
plt.show()
```

# Step 10 — Hyperparameter Tuning

### Key hyperparameters for Random Forest:
- **`n_estimators`** → number of trees (more = better but slower).
- **`max_depth`** → maximum depth of each tree (control overfitting).
- **`min_samples_split`** → minimum samples to split a node.
- **`max_features`** → number of features considered at each split (`'sqrt'` is default and usually best).

**We use `GridSearchCV` to find the best combination!**

> ⚠️ **Note:** Random Forest tuning can be SLOW! We use a small grid here to keep it fast.

```python
# ================================================================
# HYPERPARAMETER TUNING with GridSearchCV
# ================================================================
param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5],
    'max_features': ['sqrt', 'log2']
}

grid = GridSearchCV(
    RandomForestClassifier(random_state=42, n_jobs=-1),
    param_grid,
    cv=3,                # 3-fold CV (faster; 5-fold takes much longer)
    scoring='f1',         # optimize F1 (best for imbalanced churn data)
    n_jobs=-1,
    verbose=1
)

print("⏳ Tuning is running... this may take a couple of minutes.")
grid.fit(X_train, y_train)

print("\n✅ Best hyperparameters found:")
print(grid.best_params_)
print(f"Best cross-validation F1 score: {grid.best_score_:.4f}")

# Use the best model
best_model = grid.best_estimator_
y_pred_tuned = best_model.predict(X_test)
y_proba_tuned = best_model.predict_proba(X_test)[:, 1]

print("\n--- Tuned Model on Test Data ---")
print(f"Accuracy  : {accuracy_score(y_test, y_pred_tuned):.4f}")
print(f"Precision : {precision_score(y_test, y_pred_tuned):.4f}")
print(f"Recall    : {recall_score(y_test, y_pred_tuned):.4f}")
print(f"F1 Score  : {f1_score(y_test, y_pred_tuned):.4f}")
print(f"ROC-AUC   : {roc_auc_score(y_test, y_proba_tuned):.4f}")
```

```python
# ================================================================
# FINAL COMPARISON: Baseline vs Tuned Random Forest
# ================================================================
comparison_df = pd.DataFrame({
    'Model': ['Baseline Random Forest', 'Tuned Random Forest'],
    'Accuracy':  [accuracy_score(y_test, y_pred),   accuracy_score(y_test, y_pred_tuned)],
    'Precision': [precision_score(y_test, y_pred),  precision_score(y_test, y_pred_tuned)],
    'Recall':    [recall_score(y_test, y_pred),     recall_score(y_test, y_pred_tuned)],
    'F1 Score':  [f1_score(y_test, y_pred),         f1_score(y_test, y_pred_tuned)],
    'ROC-AUC':   [roc_auc_score(y_test, y_proba),   roc_auc_score(y_test, y_proba_tuned)]
})

print("=" * 75)
print("FINAL MODEL COMPARISON")
print("=" * 75)
comparison_df.round(4)
```

## Step 11 — 🎯 Feature Importance (The BIG advantage of Random Forest!)

One of the coolest things about Random Forest is that it tells us **which features matter most** for predicting churn.

This is a **HUGE business insight** — the telecom company can act on this to reduce churn!

```python
# ================================================================
# FEATURE IMPORTANCE ANALYSIS
# ================================================================
importance_df = pd.DataFrame({
    'Feature': X.columns,
    'Importance': best_model.feature_importances_
}).sort_values(by='Importance', ascending=False)

# Show top 15 most important features
top15 = importance_df.head(15)

plt.figure(figsize=(10, 7))
sns.barplot(data=top15, x='Importance', y='Feature', palette='viridis')
plt.title("Top 15 Features That Drive Customer Churn")
plt.xlabel("Importance Score")
plt.show()

print("Top 15 most important features:")
print(top15.to_string(index=False))
```

## Step 12 — Save the Final Model

Save the trained model so we can reuse it later without retraining.

```python
# ================================================================
# SAVE THE FINAL MODEL
# ================================================================
final_model = best_model   # use the tuned model

joblib.dump(final_model, "random_forest_churn_model.pkl")

# Save the list of feature names too (very useful for predicting new samples)
joblib.dump(list(X.columns), "churn_feature_names.pkl")

print("✅ Model saved as 'random_forest_churn_model.pkl'")
print("✅ Feature names saved as 'churn_feature_names.pkl'")
```

```python
# ================================================================
# BONUS: Load the saved model and predict for a NEW customer
# ================================================================
loaded_model = joblib.load("random_forest_churn_model.pkl")
feature_names = joblib.load("churn_feature_names.pkl")

# Build a new customer using the SAME feature columns as X (order matters!)
# Easiest way: take one row from X_test as a template and modify values
new_customer = X_test.iloc[[0]].copy()   # copy one existing customer row as template

# Modify a few key fields for the new customer
new_customer['tenure'] = 2                  # very new customer
new_customer['MonthlyCharges'] = 85.0        # high monthly bill
new_customer['TotalCharges'] = 170.0         # tenure * monthly ≈

predicted_class = loaded_model.predict(new_customer)[0]
predicted_proba = loaded_model.predict_proba(new_customer)[0][1]

print("=" * 45)
print("   NEW CUSTOMER CHURN PREDICTION")
print("=" * 45)
print(f"Prediction: {'🔴 Will Churn' if predicted_class == 1 else '🟢 Will Stay'}")
print(f"Churn probability: {predicted_proba*100:.1f}%")
print("=" * 45)

if predicted_proba > 0.5:
    print("\n💡 Business Action: Send this customer a retention offer!")
else:
    print("\n💡 Business Action: No action needed — customer is likely loyal.")
```

### 🏆 Golden Rules for Random Forest
1. **NO scaling needed** — trees split on raw values.
2. **Always use `stratify=y`** for classification splits.
3. **Set `n_jobs=-1`** to use all CPU cores → much faster training!
4. **`n_estimators`** — more trees = better accuracy (up to a point) but slower.
5. **Random Forest handles imbalanced data OK** but F1/Recall are more meaningful than Accuracy.
6. **Feature importance** is one of RF's biggest strengths — use it for business decisions!

### 🌟 Pros & Cons of Random Forest

| ✅ Pros | ❌ Cons |
|---|---|
| Excellent accuracy on tabular data | Slower to train than single tree |
| No scaling required | Less interpretable (100 trees vs 1) |
| Handles missing values decently | Uses more memory |
| Reduces overfitting vs single tree | Not the fastest at prediction |
| Great feature importance ranking | Boosting models (XGBoost) often beat it |
| Works well with imbalanced data | Can bias toward features with more categories |
