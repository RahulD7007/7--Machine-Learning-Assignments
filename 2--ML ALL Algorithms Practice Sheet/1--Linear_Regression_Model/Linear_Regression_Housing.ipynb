### What is Linear Regression?
Linear Regression tries to draw the **best-fit straight line** (or plane, with many features) through the data, so that it can predict a **continuous number** (like house price) instead of a category.

## What is it?

**Real-life analogy:** Imagine plotting house sizes on a graph and drawing a ruler line through the dots. If a new house is 2,000 sq ft, you follow the line up to guess its price. That's Linear Regression!

### Main idea
Find a mathematical relationship (a straight line) that best fits the training data:

$$y = w_1 x_1 + w_2 x_2 + \dots + w_n x_n + b$$

Where:
- $y$ = predicted house price
- $w_1, w_2, \dots$ = weights (how much each feature matters)
- $x_1, x_2, \dots$ = features (income, age, rooms, etc.)
- $b$ = intercept (base price when all features are 0)

# Step 4 — Data Preprocessing

### Tasks to do:
1. **Fill missing `total_bedrooms`** with the median (207 missing values).
2. **One-Hot Encode `ocean_proximity`** — it's text, and computers only understand numbers!

Good news: only 1 column has missing values and only 1 column is text.

```python
# Final check: no missing values, all numeric
print("Any missing values left?", df.isnull().sum().sum())
print("\nAll data types:")
print(df.dtypes)
```

## Step 5 — Separate Features (X) & Target (y), then Train-Test Split

- `X` = all the **clues** (input columns).
- `y` = the **answer** (`median_house_value` — a NUMBER!).

### ⚠️ IMPORTANT: NO `stratify=y` for REGRESSION!
`stratify` is only for classification (to balance categorical classes). Since we're predicting a continuous number, we skip it.

```python
# ================================================================
# FEATURES (X), TARGET (y), TRAIN-TEST SPLIT
# ================================================================
X = df.drop('median_house_value', axis=1)   # everything EXCEPT price
y = df['median_house_value']                 # what we want to predict

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,       # 20% for testing
    random_state=42        # reproducible split
    # ❌ NO stratify for regression!
)

print(f"Training neighborhoods: {X_train.shape[0]}")
print(f"Testing neighborhoods : {X_test.shape[0]}")
print(f"Total features       : {X_train.shape[1]}")
print(f"\nTarget range:")
print(f"  Training: ${y_train.min():,.0f} to ${y_train.max():,.0f}")
print(f"  Testing : ${y_test.min():,.0f} to ${y_test.max():,.0f}")
```

# Step 6 — Feature Scaling ✅ RECOMMENDED

### 💡 *Why do we scale?*
Look at our features:
- `population` → ranges from 3 to 35,000 (HUGE numbers)
- `median_income` → ranges from 0.5 to 15 (tiny numbers)

Without scaling, the model might think `population` is 1,000× more important just because its numbers are bigger! Scaling makes all features equally important.

**StandardScaler** brings every feature to $\mu = 0$, $\sigma = 1$:
$$z = \frac{x - \mu}{\sigma}$$

### ⚠️ Golden Rule: `fit_transform()` on TRAIN, only `transform()` on TEST!

```python
# ================================================================
# FEATURE SCALING
# ================================================================
scaler = StandardScaler()

# Learn mean & std from TRAINING data only, then scale it
X_train_scaled = scaler.fit_transform(X_train)

# Apply SAME scaling to TEST data (never re-learn from test!)
X_test_scaled = scaler.transform(X_test)

print("✅ Feature scaling done!")
print(f"Example row BEFORE scaling: {X_train.iloc[0].values[:4].round(2)}")
print(f"Example row AFTER scaling : {X_train_scaled[0][:4].round(2)}")
```

# Step 7 — Model Training

### 💡 *How Linear Regression learns:*
1. Start with random weights ($w_1, w_2, \dots$) and bias ($b$).
2. Make predictions using the formula: $y = w_1 x_1 + w_2 x_2 + \dots + b$
3. Calculate the **error** (difference between predicted and actual prices).
4. Adjust weights to **minimize the total error** (using Ordinary Least Squares).
5. Repeat until the best-fit line is found!

### Key parameters:
- **`fit_intercept=True`** (default) → includes the bias term $b$.

```python
# ================================================================
# MODEL TRAINING
# ================================================================
model = LinearRegression()

# .fit() finds the best-fit line through all the data points
model.fit(X_train_scaled, y_train)

print("✅ Linear Regression model trained!")
print(f"Intercept (base price): ${model.intercept_:,.0f}")
print("\nHow much each clue affects the price (coefficients):")

coef_df = pd.DataFrame({
    'Feature': X.columns,
    'Weight': model.coef_
}).sort_values(by='Weight', key=abs, ascending=False)

for _, row in coef_df.iterrows():
    direction = "↑ raises price" if row['Weight'] > 0 else "↓ lowers price"
    print(f"  {row['Feature']:28s} -> ${row['Weight']:>12,.0f}  ({direction})")
```

# Step 8 — Prediction

For regression, we only need `.predict()` — no probabilities like in classification!

```python
# ================================================================
# PREDICTION
# ================================================================
y_pred = model.predict(X_test_scaled)

# Compare actual vs predicted prices for first 10 test neighborhoods
comparison = pd.DataFrame({
    'Actual Price ($)': y_test.values[:10].round(0),
    'Predicted Price ($)': y_pred[:10].round(0)
})
comparison['Error ($)'] = (comparison['Actual Price ($)'] - comparison['Predicted Price ($)']).round(0)
comparison['Error (%)'] = (abs(comparison['Error ($)']) / comparison['Actual Price ($)'] * 100).round(1)
comparison
```

# Step 9 — Evaluate the Model (REGRESSION METRICS!)

⚠️ For regression, we use **completely different metrics** than classification:

| Metric | Formula | Meaning | What's good? |
|---|---|---|---|
| **MAE** | $\frac{1}{n}\sum|y - \hat{y}|$ | On average, how far off in dollars? | Lower = better |
| **MSE** | $\frac{1}{n}\sum(y - \hat{y})^2$ | Penalizes big mistakes more | Lower = better |
| **RMSE** | $\sqrt{MSE}$ | Same units as target ($) | Lower = better |
| **R² Score** | $1 - \frac{\text{residual variance}}{\text{total variance}}$ | How much variance we explain (0-1) | **Closer to 1 = better** |

### 🎯 R² Score Interpretation:
- **R² = 1.0** → Perfect predictions! 🎯
- **R² = 0.90+** → Excellent (Grade A+) ✨
- **R² = 0.80** → Good (Grade A)
- **R² = 0.60** → Okay (Grade B)
- **R² = 0.0** → Model is no better than guessing the average! 😞

```python
# ================================================================
# MODEL EVALUATION
# ================================================================
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("=" * 55)
print("   BASELINE LINEAR REGRESSION — REPORT CARD")
print("=" * 55)
print(f"  MAE  (Mean Absolute Error) : ${mae:,.0f}")
print(f"  MSE  (Mean Squared Error)  : ${mse:,.0f}")
print(f"  RMSE (Root Mean Sq. Error) : ${rmse:,.0f}")
print(f"  R² Score                    : {r2:.4f} ({r2*100:.1f}% variance explained)")
print("=" * 55)

print(f"\n💡 Interpretation:")
print(f"  → On average, our predictions are off by ~${mae:,.0f}")
print(f"  → Our model explains {r2*100:.1f}% of the price variation")
if r2 > 0.8:
    print(f"  → ✅ Good model! R² > 0.80")
elif r2 > 0.6:
    print(f"  → 🤔 Decent model, but room for improvement.")
else:
    print(f"  → ⚠️ Model needs significant improvement.")
```

```python
# Actual vs Predicted plot — perfect predictions would lie on the red diagonal
plt.figure(figsize=(7, 6))
plt.scatter(y_test, y_pred, alpha=0.3, color='teal', edgecolor='black')
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()],
         'r--', lw=2, label='Perfect prediction')
plt.xlabel('Actual Price ($)')
plt.ylabel('Predicted Price ($)')
plt.title('Actual vs Predicted House Prices')
plt.legend()
plt.show()

print("💡 Points close to the red line = accurate predictions.")
print("   Points far from the line = errors.")
```

```python
# Residuals plot — errors should be randomly scattered around 0
residuals = y_test - y_pred

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

# Residuals vs Predicted
axes[0].scatter(y_pred, residuals, alpha=0.3, color='purple', edgecolor='black')
axes[0].axhline(y=0, color='red', linestyle='--', lw=2)
axes[0].set_xlabel('Predicted Price ($)')
axes[0].set_ylabel('Residuals (Actual - Predicted)')
axes[0].set_title('Residuals vs Predicted')

# Residuals distribution
sns.histplot(residuals, kde=True, color='salmon', ax=axes[1])
axes[1].set_xlabel('Residuals ($)')
axes[1].set_title('Distribution of Residuals')
axes[1].axvline(x=0, color='red', linestyle='--', lw=2)

plt.tight_layout()
plt.show()

print("💡 Good signs:")
print("  - Residuals randomly scattered around 0 (no pattern)")
print("  - Residuals normally distributed (bell curve shape)")
```

# Step 10 — Hyperparameter Tuning (Ridge & Lasso)

### 💡 *Why do we need regularization?*
Sometimes a model "memorizes" the training data too closely (**overfitting**). Regularization adds a penalty so the model stays simple.

| Type | Formula | Effect |
|---|---|---|
| **Ridge (L2)** | Loss = MSE + $\alpha \sum w_i^2$ | Shrinks weights smoothly toward zero |
| **Lasso (L1)** | Loss = MSE + $\alpha \sum |w_i|$ | Can shrink useless weights to **exactly 0** (feature selection!) |

- **`alpha`** is the "strictness knob" — bigger alpha = stronger penalty.
- **`GridSearchCV`** tries several alpha values and picks the best one!

```python
# ================================================================
# RIDGE REGRESSION (L2 Regularization)
# ================================================================
ridge_params = {'alpha': [0.001, 0.01, 0.1, 1.0, 10.0, 100.0]}
ridge_grid = GridSearchCV(Ridge(), ridge_params, cv=5, scoring='r2')
ridge_grid.fit(X_train_scaled, y_train)

best_ridge = ridge_grid.best_estimator_
y_pred_ridge = best_ridge.predict(X_test_scaled)
r2_ridge = r2_score(y_test, y_pred_ridge)
rmse_ridge = np.sqrt(mean_squared_error(y_test, y_pred_ridge))

print("--- Ridge Regression Results ---")
print(f"Best Alpha : {ridge_grid.best_params_['alpha']}")
print(f"R² Score   : {r2_ridge:.4f}")
print(f"RMSE       : ${rmse_ridge:,.0f}")
```

```python
# ================================================================
# LASSO REGRESSION (L1 Regularization)
# ================================================================
lasso_params = {'alpha': [0.0001, 0.001, 0.01, 0.1, 1.0]}
lasso_grid = GridSearchCV(Lasso(max_iter=5000), lasso_params, cv=5, scoring='r2')
lasso_grid.fit(X_train_scaled, y_train)

best_lasso = lasso_grid.best_estimator_
y_pred_lasso = best_lasso.predict(X_test_scaled)
r2_lasso = r2_score(y_test, y_pred_lasso)
rmse_lasso = np.sqrt(mean_squared_error(y_test, y_pred_lasso))

print("--- Lasso Regression Results ---")
print(f"Best Alpha             : {lasso_grid.best_params_['alpha']}")
print(f"R² Score               : {r2_lasso:.4f}")
print(f"RMSE                   : ${rmse_lasso:,.0f}")
print(f"Features eliminated to 0: {np.sum(best_lasso.coef_ == 0)}")
```

```python
# ================================================================
# FINAL COMPARISON: OLS vs Ridge vs Lasso
# ================================================================
comparison_df = pd.DataFrame({
    'Model': ['Linear Regression (OLS)', 'Ridge Regression (L2)', 'Lasso Regression (L1)'],
    'Best Alpha': ['None', ridge_grid.best_params_['alpha'], lasso_grid.best_params_['alpha']],
    'R² Score': [r2, r2_ridge, r2_lasso],
    'RMSE ($)': [round(rmse), round(rmse_ridge), round(rmse_lasso)]
})

print("=" * 65)
print("FINAL MODEL COMPARISON")
print("=" * 65)
comparison_df
```

### 🏆 Golden Rules for Linear Regression
1. **NO stratify for regression** — only for classification.
2. **Scaling is recommended** — especially when using Ridge/Lasso.
3. **`fit_transform()` on train, only `transform()` on test** — prevent data leakage!
4. **R² close to 1.0 is good**, but always check MAE/RMSE in real dollars too.
5. **Ridge (L2)** shrinks weights smoothly; **Lasso (L1)** can eliminate useless features.
6. **Regression metrics ≠ Classification metrics!** Use MAE, RMSE, R² — not accuracy.
7. **Always save the scaler** — you need it to predict new samples correctly!

### 🌟 Pros & Cons of Linear Regression

| ✅ Pros | ❌ Cons |
|---|---|
| Very simple and interpretable | Assumes linear relationship |
| Extremely fast to train | Sensitive to outliers |
| Coefficients show feature importance | Struggles with complex patterns |
| Great baseline model | Assumes no multicollinearity |
| Works well on small datasets | Can underfit non-linear data |
| Foundation for Ridge, Lasso, ElasticNet | Assumes normally distributed residuals |
