# ML Workflow — Step-by-Step Beginner Guide

**Audience:** A 14-year-old college student / complete beginner

## The Big Picture

Machine Learning is like teaching a computer with examples.

**Import → Load Data → EDA → Preprocess → Split → Scale if needed → Train → Predict → Evaluate → Tune**

### Simple analogy

Imagine teaching a student to identify fruits:

1. Collect examples.
2. Understand the examples.
3. Clean incorrect information.
4. Keep some examples hidden for the final test.
5. Teach using the training examples.
6. Give unseen examples.
7. Check the answers.
8. Improve the teaching method.

> **Golden Rule:** The overall workflow is similar, but each algorithm has different strengths, preprocessing needs, parameters, and evaluation choices.

# 1. Import Libraries

Libraries are ready-made tools.

```python

    silhouette_score

```

**Simple explanation:** Instead of writing every mathematical operation ourselves, we import tools that already know how to work with data and ML models.


# 2. Load Dataset

For a CSV file:

```python
df = pd.read_csv("your_dataset.csv")

print(df.head())
```

- `df` means DataFrame.
- A DataFrame looks like a table.
- Rows are examples/records.
- Columns are features or the target.


# 3. EDA — Exploratory Data Analysis

EDA means **understanding the data before training**.

Useful commands:

```python
df.head()
df.shape
df.info()
df.describe()
df.isnull().sum()
df.duplicated().sum()
```

Useful questions:

- How many rows and columns?
- Which columns are numeric?
- Are values missing?
- Are there duplicates?
- Are there strange/outlier values?
- Is the target balanced?
- Which features seem related?

Example graphs:

```python
sns.histplot(df["feature"])
plt.show()

sns.boxplot(x=df["feature"])
plt.show()

sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.show()
```

> **Golden Rule:** Understand the data before asking a model to learn from it.


# 4. Data Preprocessing

Preprocessing means **making the data ready for ML**.

Typical tasks:

- Remove or handle duplicates.
- Handle missing values.
- Convert text categories into numbers.
- Detect unreasonable values/outliers.
- Select useful features.
- Separate features (`X`) and target (`y`).

Example:

```python
# Features
X = df.drop(columns=["target"])

# Target
y = df["target"]
```

For categorical data, common choices are one-hot encoding or ordinal encoding, depending on the meaning of the category.

> **Important:** Do not fit preprocessing steps on the complete dataset before splitting. Fit them on training data and apply the learned transformation to test data. This helps prevent data leakage.


# 5. Train-Test Split

We need a fair exam.

- **Training data** → model learns from it.
- **Testing data** → model has never seen it; we use it to check performance.

Typical split:

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)
```

For classification, often use:

```python
train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

`random_state=42` makes the split reproducible.


# 6. Feature Scaling

Scaling puts numeric features onto comparable ranges.

### Example

If:

- Age = 20–60
- Salary = 20,000–200,000

salary has much larger numerical values.

Distance-based algorithms can be affected by this.

### Common scaling

```python
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

### Usually important

- KNN
- SVM
- K-Means
- Logistic Regression

### Usually unnecessary for tree-based models

- Decision Tree
- Random Forest
- Gradient Boosting
- AdaBoost
- XGBoost

Basic Linear Regression usually does not require scaling, although scaling can be useful in some variants/workflows.

**Golden Rule:** Never `fit_transform()` the test data. Learn scaling parameters from training data only.

# 9. Model Evaluation

## Regression

- MAE → average absolute error; lower is better.
- RMSE → penalizes larger errors more; lower is better.
- R² → how much variation is explained; higher is generally better.

## Classification

- Accuracy → percentage of correct predictions.
- Precision → among predicted positives, how many were correct?
- Recall → among actual positives, how many were found?
- F1 → balance of precision and recall.
- Confusion Matrix → detailed count of correct/wrong class predictions.

**Important:** Accuracy is not always the best metric, especially when classes are highly imbalanced.


# 10. Hyperparameter Tuning

Examples:

- Random Forest → `n_estimators`, `max_depth`

- SVM → `C`, `gamma`, `kernel`

# 11. Linear Regression

## What is it?

Linear Regression predicts a **continuous number**.

Example:

**House features → model → predicted house price**

### Recommended dataset
**California Housing**

### Workflow
1. Load data.
2. Separate X and y.
3. Split.
4. Train.
5. Predict.
6. Evaluate with MAE, RMSE, R².

### Main idea
Find a mathematical relationship that best fits the training data.


````python
from sklearn.datasets import fetch_california_housing
from sklearn.linear_model import LinearRegression

# 1. Load a built-in regression dataset
data = fetch_california_housing(as_frame=True)
df = data.frame

# 2. Features and target
X = df.drop(columns=["MedHouseVal"])
y = df["MedHouseVal"]

# 3. Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

# 4. Train
model = LinearRegression()
model.fit(X_train, y_train)

# 5. Predict
y_pred = model.predict(X_test)

# 6. Evaluate
print("MAE :", mean_absolute_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R2  :", r2_score(y_test, y_pred))
````

## Linear Regression — Remember

- Type: **Regression**
- Output: **Number**
- Scaling: Usually optional for basic Linear Regression
- Metrics: **MAE, RMSE, R²**
- Easy sentence: **"Linear Regression predicts a number by learning a best-fit relationship between inputs and output."**


# 12. Logistic Regression

## What is it?

Despite its name, Logistic Regression is mainly a **classification** algorithm.

Example:

**Medical measurements → model → Benign / Malignant**

### Recommended dataset
**Breast Cancer Wisconsin**

### Scaling
Recommended because it is sensitive to feature scale.


````python
from sklearn.datasets import load_breast_cancer
from sklearn.linear_model import LogisticRegression

# 1. Load data
data = load_breast_cancer(as_frame=True)
X = data.data
y = data.target

# 2. Split while keeping class proportions similar
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# 3. Scale
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 4. Train
model = LogisticRegression(max_iter=1000)
model.fit(X_train_scaled, y_train)

# 5. Predict
y_pred = model.predict(X_test_scaled)

# 6. Evaluate
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
````

## Logistic Regression — Remember

- Type: **Classification**
- Output: **Class/probability**
- Scaling: Recommended
- Metrics: Accuracy, Precision, Recall, F1, ROC-AUC
- Easy sentence: **"Logistic Regression estimates the probability of a class and uses it to make a classification decision."**


# 13. Decision Tree

## What is it?

A Decision Tree makes predictions by asking a sequence of questions.

Example:

**Age < 30? → Yes → Income high? → Yes → Class A**

### Recommended dataset
**Titanic** for realistic practice; **Iris** for an easy first example.

### Scaling
Not required.

### Important parameters
- `max_depth`
- `min_samples_split`
- `min_samples_leaf`
- `criterion`

### Main danger
A very deep tree can memorize training data → **overfitting**.


````python
from sklearn.tree import DecisionTreeClassifier

# Use the breast-cancer X_train/X_test/y_train/y_test from above.
# Trees do not need feature scaling.

model = DecisionTreeClassifier(
    max_depth=4,
    random_state=42
)

model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
````

# 14. Random Forest

## What is it?

Random Forest = **many Decision Trees + combined prediction**.

Analogy:

- One tree = one student.
- Many trees = many students.
- Forest = combine their answers.

### Recommended dataset
Titanic or Breast Cancer.

### Scaling
Not required.

### Important parameters
- `n_estimators` → number of trees
- `max_depth`
- `max_features`
- `min_samples_leaf`


````python
from sklearn.ensemble import RandomForestClassifier

model = RandomForestClassifier(
    n_estimators=100,
    max_depth=6,
    random_state=42
)

model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
````

# 15. Support Vector Machine (SVM)

## What is it?

SVM tries to find a strong boundary that separates classes.

Imagine red balls and blue balls. SVM tries to draw a separating line/plane with a good margin.

### Recommended dataset
Breast Cancer Wisconsin.

### Scaling
**Strongly recommended** for typical SVM workflows.

### Important parameters
- `C` → controls the trade-off between margin width and classification errors.
- `kernel` → type/shape of decision boundary.
- `gamma` → controls influence of points for kernels such as RBF.


````python
from sklearn.svm import SVC

# Scale first
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Train SVM
model = SVC(
    kernel="rbf",
    C=1.0,
    gamma="scale"
)
model.fit(X_train_scaled, y_train)

# Predict and evaluate
y_pred = model.predict(X_test_scaled)

print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
````

# 16. K-Nearest Neighbors (KNN)

## What is it?

KNN looks at the **nearest examples** and uses them to decide the class.

Example:

If most of the 5 nearest points belong to Class A, the new point is likely Class A.

### Recommended dataset
**Iris**

### Scaling
**Very important**, because KNN uses distances.

### Important parameter
`n_neighbors` = K = number of nearby points considered.


````python
from sklearn.neighbors import KNeighborsClassifier

# Scale because KNN is distance-based
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model = KNeighborsClassifier(n_neighbors=5)
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)

print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
````

# 18. Ensemble Learning

## What is an ensemble?

Ensemble learning means **combining multiple models to create a stronger solution**.

Analogy:

Instead of asking one student, ask 10 students and combine their answers.

### Two important families

**Bagging**
- Models are trained independently.
- Their predictions are combined.
- Example: Random Forest.

**Boosting**
- Models are built sequentially.
- Later models focus on mistakes made earlier.
- Examples: AdaBoost, Gradient Boosting, XGBoost.

### Memory sentence
**Ensemble = multiple learners + smart combination.**


# 19. Gradient Boosting

## What is it?

Gradient Boosting builds models **one after another**.

Each new model tries to improve the errors left by previous models.

**Model 1 → mistakes → Model 2 improves → Model 3 improves → final prediction**

### Recommended datasets
- California Housing → regression.
- Titanic/Breast Cancer → classification.

### Scaling
Usually not required for tree-based Gradient Boosting.

### Important parameters
- `n_estimators`
- `learning_rate`
- `max_depth`


````python
from sklearn.ensemble import GradientBoostingClassifier

model = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.05,
    max_depth=3,
    random_state=42
)

model.fit(X_train, y_train)
y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
````

# 23. DBSCAN

## What is it?

DBSCAN groups points according to **density**.

Unlike K-Means:

- You do not need to choose K.
- It can find irregularly shaped clusters.
- It can mark isolated observations as **noise**.

### Recommended first practice
`make_moons` is excellent for visual understanding.

### Scaling
Usually important when features have different scales.

### Important parameters
- `eps` → neighborhood radius.
- `min_samples` → minimum points needed for a dense region.


````python
from sklearn.datasets import make_moons
from sklearn.cluster import DBSCAN

# Create a simple curved dataset
X, y_true = make_moons(
    n_samples=300,
    noise=0.08,
    random_state=42
)

# Scale
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Train DBSCAN
dbscan = DBSCAN(
    eps=0.25,
    min_samples=5
)

labels = dbscan.fit_predict(X_scaled)

print("Clusters/noise labels:", np.unique(labels))
print("Noise points:", np.sum(labels == -1))

# Visualize
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=labels)
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("DBSCAN Clusters")
plt.show()
````

# 28. One-Line Explanation of Every Algorithm

- **Linear Regression:** Predict a number using a best-fit relationship.
- **Logistic Regression:** Predict the probability/class of an example.
- **Decision Tree:** Make predictions using a sequence of questions.
- **Random Forest:** Combine many decision trees.
- **SVM:** Find a strong separating boundary with a useful margin.
- **KNN:** Look at nearby examples and vote.
- **Naive Bayes:** Use probability to choose the most likely class.
- **Ensemble Learning:** Combine multiple models.
- **Gradient Boosting:** Build models sequentially to reduce previous errors.
- **AdaBoost:** Give more attention to examples previous learners got wrong.
- **XGBoost:** A powerful, regularized implementation of gradient boosting.
- **K-Means:** Divide observations into K groups around cluster centers.
- **DBSCAN:** Find dense groups and identify noise.
