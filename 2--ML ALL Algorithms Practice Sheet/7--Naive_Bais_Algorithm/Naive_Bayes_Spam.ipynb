### What is Naive Bayes?
Naive Bayes is a **probability-based** algorithm that uses **Bayes' Theorem** to calculate the chance that a message belongs to a class (spam or ham).

## What is it?

**Real-life analogy:** Imagine you get a message that says "FREE MONEY NOW!!!". Your brain instantly thinks:
> "Hmm, the words 'FREE', 'MONEY', and '!!!' appear in spam messages 90% of the time. So this is probably spam!"

That's exactly what Naive Bayes does — it counts how often each word appears in spam vs ham messages and calculates the probability!

### Main idea — Bayes' Theorem:
$$P(\text{Spam} | \text{message}) = \frac{P(\text{message} | \text{Spam}) \times P(\text{Spam})}{P(\text{message})}$$

In simple words:
> **"Given these words, what's the probability this message is spam?"**

### Why "Naive"?
It assumes all words are **independent** of each other (which is unrealistic — "free" and "money" often appear together). But despite this "naive" assumption, it works **surprisingly well** for text classification!

# Step 4 — Text Preprocessing (THE MOST IMPORTANT STEP!)

### 💡 *Why do we preprocess text?*
Raw text is messy! Computers need clean, standardized input. We do 4 things:

1. **Lowercase** → "FREE" and "free" should be treated as the same word.
2. **Remove special characters & numbers** → "Win $1000!!!" → "win".
3. **Remove stopwords** → Words like "the", "is", "at", "and" appear everywhere and carry no meaning.
4. **Stemming** → Reduce words to their root: "running" → "run", "loved" → "love", "winner" → "winner".

> **Golden Rule of NLP:** Garbage in = Garbage out. Better preprocessing = Better model!

```python
# ================================================================
# TEXT PREPROCESSING FUNCTION
# ================================================================

# Initialize tools
stemmer = PorterStemmer()
stop_words = set(stopwords.words('english'))

def clean_text(text):
    """
    Cleans a single SMS message through 4 steps:
    1. Lowercase
    2. Remove special characters & numbers
    3. Remove stopwords
    4. Stemming
    """
    # Step 1: Convert to lowercase
    text = text.lower()
    
    # Step 2: Remove everything except letters (no numbers, no punctuation)
    text = re.sub(r'[^a-zA-Z]', ' ', text)
    
    # Step 3 & 4: Split into words, remove stopwords, and stem
    words = text.split()
    words = [stemmer.stem(word) for word in words if word not in stop_words]
    
    # Join back into a single string
    return ' '.join(words)

# Apply cleaning to all messages
print("⏳ Cleaning all messages... (this takes a few seconds)")
df['clean_message'] = df['message'].apply(clean_text)
print("✅ Text preprocessing complete!")

# Show before vs after
print("\n" + "=" * 70)
print("BEFORE vs AFTER CLEANING (5 examples):")
print("=" * 70)
for i in range(5):
    print(f"\n📩 BEFORE: {df['message'].iloc[i][:80]}")
    print(f"✨ AFTER : {df['clean_message'].iloc[i][:80]}")
```

```python
# Encode target: spam=1, ham=0
df['label_encoded'] = df['label'].map({'ham': 0, 'spam': 1})

print("✅ Label encoded (ham=0, spam=1)")
print(df['label_encoded'].value_counts())
```

# Step 5 — Feature Extraction (Vectorization)

### 💡 *Why do we need this?*
Machine Learning models can ONLY work with **numbers**, not text!
We need to convert our cleaned messages into a **numerical matrix**.

### Two popular methods:

| Method | How it works | Best for |
|---|---|---|
| **CountVectorizer** | Counts how many times each word appears | Simple text |
| **TF-IDF Vectorizer** ✅ | Counts word frequency BUT reduces weight of common words | **Better for spam detection!** |

### TF-IDF = Term Frequency × Inverse Document Frequency
- **TF** → How often does this word appear in THIS message?
- **IDF** → How rare is this word across ALL messages? (rare words get higher scores!)

Example: The word "call" appears in many spam messages but rarely in ham → HIGH TF-IDF score → strong spam indicator!

```python
# ================================================================
# FEATURE EXTRACTION — Convert text to numbers using TF-IDF
# ================================================================

vectorizer = TfidfVectorizer(
    max_features=3000,      # keep only the top 3000 most important words
    ngram_range=(1, 2),      # use single words AND 2-word phrases (e.g., 'free call')
    min_df=2,                # ignore words that appear in fewer than 2 messages
    max_df=0.95              # ignore words that appear in more than 95% of messages
)

# Fit and transform the cleaned messages into a numerical matrix
X = vectorizer.fit_transform(df['clean_message'])
y = df['label_encoded']

print(f"✅ Vectorization complete!")
print(f"Number of messages : {X.shape[0]}")
print(f"Number of features : {X.shape[1]} (top words/phrases)")
print(f"Matrix type        : {type(X).__name__} (sparse matrix — memory efficient!)")

# Show some of the vocabulary words the vectorizer learned
vocab = vectorizer.get_feature_names_out()
print(f"\nSample vocabulary words: {list(vocab[:20])}")
```

## Step 6 — Train-Test Split

### ⚠️ IMPORTANT: Use `stratify=y`
Our data is imbalanced (87% ham, 13% spam), so stratify ensures both train and test have the same ratio!

```python
# ================================================================
# TRAIN-TEST SPLIT
# ================================================================
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,       # 20% for testing
    random_state=42,       # reproducible split
    stratify=y             # ✅ preserve ham/spam ratio
)

print(f"Training messages: {X_train.shape[0]}")
print(f"Testing messages : {X_test.shape[0]}")
print(f"Total features   : {X_train.shape[1]}")
print("\nTrain class balance:")
print(f"  Ham  (0): {(y_train == 0).sum()} ({(y_train == 0).mean()*100:.1f}%)")
print(f"  Spam (1): {(y_train == 1).sum()} ({(y_train == 1).mean()*100:.1f}%)")
print("\nTest class balance:")
print(f"  Ham  (0): {(y_test == 0).sum()} ({(y_test == 0).mean()*100:.1f}%)")
print(f"  Spam (1): {(y_test == 1).sum()} ({(y_test == 1).mean()*100:.1f}%)")
```

# Step 7 — Model Training

### 💡 *How Naive Bayes learns:*
1. For each word in the vocabulary, calculate:
   - P(word | spam) → How likely is this word in spam messages?
   - P(word | ham) → How likely is this word in ham messages?
2. When a new message arrives, multiply the probabilities of all its words.
3. Whichever class (spam/ham) has the higher total probability wins!

### Key parameter:
- **`alpha=1.0`** → Laplace smoothing. Prevents zero probability when a word was never seen during training. Think of it as giving every word a tiny "benefit of the doubt" score.

```python
# ================================================================
# MODEL TRAINING
# ================================================================
model = MultinomialNB(alpha=1.0)   # alpha=1.0 is Laplace smoothing (default)

# .fit() calculates word probabilities for spam and ham
model.fit(X_train, y_train)

print("✅ Multinomial Naive Bayes model trained!")
print(f"Number of classes: {model.class_count_}")
print(f"  Ham  training samples: {int(model.class_count_[0])}")
print(f"  Spam training samples: {int(model.class_count_[1])}")
print(f"Number of features (words): {model.feature_count_.shape[1]}")
```

# Step 8 — Prediction

```python
# ================================================================
# PREDICTION
# ================================================================
y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:, 1]   # probability of being spam

# Compare actual vs predicted for first 10 test messages
reverse_map = {0: 'ham', 1: 'spam'}

comparison = pd.DataFrame({
    'Actual': [reverse_map[v] for v in y_test.values[:10]],
    'Predicted': [reverse_map[v] for v in y_pred[:10]],
    'Spam Probability': y_proba[:10].round(3)
})
comparison['Correct?'] = comparison['Actual'] == comparison['Predicted']
comparison
```

# Step 9 — Evaluate the Model

| Metric | Meaning |
|---|---|
| **Accuracy** | Overall correctness |
| **Precision** | Of predicted spam, how many were actually spam? |
| **Recall** | Of actual spam, how many did we catch? |
| **F1 Score** | Balance of Precision & Recall |

### ⚠️ In spam detection, high **PRECISION** is critical!
A False Positive (marking a real message as spam) means someone might miss an important text from their boss or family! That's much worse than letting a spam message through (False Negative).

```python
# ================================================================
# MODEL EVALUATION
# ================================================================
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("=" * 50)
print("   BASELINE NAIVE BAYES — REPORT CARD")
print("=" * 50)
print(f"  Accuracy  : {acc:.4f} ({acc*100:.1f}%)")
print(f"  Precision : {prec:.4f}")
print(f"  Recall    : {rec:.4f}")
print(f"  F1 Score  : {f1:.4f}")
print("=" * 50)

print("\n📋 Detailed Classification Report:")
print(classification_report(y_test, y_pred, target_names=['Ham (0)', 'Spam (1)']))
```

```python
# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Pred Ham', 'Pred Spam'],
            yticklabels=['Actual Ham', 'Actual Spam'])
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

tn, fp, fn, tp = cm.ravel()
print(f"✅ True Negatives  (TN): {tn}  -> Correctly said 'Ham'")
print(f"❌ False Positives (FP): {fp}  -> Marked real message as spam (BAD!)")
print(f"❌ False Negatives (FN): {fn}  -> Let spam through (less critical)")
print(f"✅ True Positives  (TP): {tp}  -> Correctly caught spam")
```

# Step 10 — Hyperparameter Tuning

### Key hyperparameters for Naive Bayes:
- **`alpha`** → Smoothing parameter. Controls how much we "smooth" word probabilities.
  - Small alpha (0.1) → less smoothing, more sensitive to rare words
  - Large alpha (10) → more smoothing, treats all words more equally
  - Default = 1.0 (Laplace smoothing)

We'll also compare **MultinomialNB** vs **BernoulliNB** to see which works better!

```python
# ================================================================
# HYPERPARAMETER TUNING with GridSearchCV
# ================================================================
param_grid = {
    'alpha': [0.01, 0.1, 0.5, 1.0, 2.0, 5.0, 10.0]
}

grid = GridSearchCV(
    MultinomialNB(),
    param_grid,
    cv=5,                # 5-fold cross-validation
    scoring='f1',         # optimize F1
    n_jobs=-1
)

grid.fit(X_train, y_train)

print("✅ Best hyperparameters found:")
print(grid.best_params_)
print(f"Best cross-validation F1 score: {grid.best_score_:.4f}")

# Use the best model
best_model = grid.best_estimator_
y_pred_tuned = best_model.predict(X_test)

print("\n--- Tuned Model on Test Data ---")
print(f"Accuracy  : {accuracy_score(y_test, y_pred_tuned):.4f}")
print(f"Precision : {precision_score(y_test, y_pred_tuned):.4f}")
print(f"Recall    : {recall_score(y_test, y_pred_tuned):.4f}")
print(f"F1 Score  : {f1_score(y_test, y_pred_tuned):.4f}")
```

```python
# ================================================================
# BONUS: Compare MultinomialNB vs BernoulliNB
# ================================================================
bnb = BernoulliNB(alpha=1.0)
bnb.fit(X_train, y_train)
y_pred_bnb = bnb.predict(X_test)

comparison_df = pd.DataFrame({
    'Model': ['MultinomialNB (Baseline)', f'MultinomialNB (Tuned, α={grid.best_params_["alpha"]})', 'BernoulliNB'],
    'Accuracy':  [accuracy_score(y_test, y_pred),
                  accuracy_score(y_test, y_pred_tuned),
                  accuracy_score(y_test, y_pred_bnb)],
    'Precision': [precision_score(y_test, y_pred),
                  precision_score(y_test, y_pred_tuned),
                  precision_score(y_test, y_pred_bnb)],
    'Recall':    [recall_score(y_test, y_pred),
                  recall_score(y_test, y_pred_tuned),
                  recall_score(y_test, y_pred_bnb)],
    'F1 Score':  [f1_score(y_test, y_pred),
                  f1_score(y_test, y_pred_tuned),
                  f1_score(y_test, y_pred_bnb)]
})

print("=" * 80)
print("FINAL MODEL COMPARISON")
print("=" * 80)
comparison_df.round(4)
```

Save the trained model **AND** the vectorizer — we need the vectorizer to convert any new message text into numbers before predicting!

```python
# ================================================================
# SAVE THE FINAL MODEL + VECTORIZER
# ================================================================
final_model = best_model   # use the tuned model

joblib.dump(final_model, "naive_bayes_spam_model.pkl")
joblib.dump(vectorizer, "spam_tfidf_vectorizer.pkl")

print("✅ Model saved as 'naive_bayes_spam_model.pkl'")
print("✅ Vectorizer saved as 'spam_tfidf_vectorizer.pkl'")
```

```python
# ================================================================
# BONUS: Load the saved model and predict for NEW messages!
# ================================================================
loaded_model = joblib.load("naive_bayes_spam_model.pkl")
loaded_vectorizer = joblib.load("spam_tfidf_vectorizer.pkl")

# Test with some real-world messages
test_messages = [
    "Hey, are we still meeting for lunch tomorrow?",
    "CONGRATULATIONS! You've won a $1000 Walmart gift card! Click here to claim NOW!",
    "Mom, please call me when you get home.",
    "FREE entry in 2 a weekly competition to win an iPhone! Text WIN to 88888 now!",
    "Can you send me the notes from today's class?",
    "URGENT: Your bank account has been compromised. Reply with your PIN to verify."
]

print("=" * 70)
print("   TESTING NEW MESSAGES WITH OUR SPAM DETECTOR")
print("=" * 70)

for msg in test_messages:
    # Step 1: Clean the message (same preprocessing as training!)
    cleaned = clean_text(msg)
    
    # Step 2: Vectorize using the SAME vectorizer
    vectorized = loaded_vectorizer.transform([cleaned])
    
    # Step 3: Predict
    pred = loaded_model.predict(vectorized)[0]
    prob = loaded_model.predict_proba(vectorized)[0][1]
    
    label = "🚨 SPAM" if pred == 1 else "📩 HAM"
    print(f"\n{label} ({prob*100:.1f}% spam)")
    print(f"  Message: \"{msg[:65]}...\"" if len(msg) > 65 else f"  Message: \"{msg}\"")

print("\n" + "=" * 70)
```

### 🏆 Golden Rules for Naive Bayes
1. **Text preprocessing is MORE important than the model itself!** Garbage in = garbage out.
2. **Always use `stratify=y`** for classification splits.
3. **NO scaling needed** — Naive Bayes works on word counts/probabilities.
4. **`MultinomialNB`** is the best variant for text classification.
5. **Always save the vectorizer** — you need it to convert new text to numbers!
6. **In spam detection, high Precision is critical** — don't mark real messages as spam!
7. **TF-IDF is better than CountVectorizer** — it gives rare but important words more weight.

### 🌟 Pros & Cons of Naive Bayes

| ✅ Pros | ❌ Cons |
|---|---|
| Extremely fast training and prediction | "Naive" assumption (word independence) is unrealistic |
| Works great with high-dimensional text data | Can't capture word order or context |
| Needs very little training data | Struggles with sarcasm and complex language |
| Handles multi-class problems naturally | Performance plateaus with more data |
| No scaling required | Less accurate than deep learning for complex NLP |
| Great baseline for any text classification | Sensitive to vocabulary changes |

### 🎯 Key NLP Concepts to Remember
1. **Tokenization** = splitting text into individual words.
2. **Stopwords** = common filler words ("the", "is", "at") that we remove.
3. **Stemming** = reducing words to root form ("running" → "run").
4. **TF-IDF** = scoring words by importance (rare words in a document = high score).
5. **Vectorization** = converting text into a numerical matrix.
6. **Laplace Smoothing (alpha)** = prevents zero probability for unseen words.
