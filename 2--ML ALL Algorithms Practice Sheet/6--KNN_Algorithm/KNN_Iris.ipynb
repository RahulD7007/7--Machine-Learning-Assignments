### What is K-Nearest Neighbors (KNN)?
KNN is the simplest ML algorithm! It works on one basic rule:

> **"Tell me who your neighbors are, and I'll tell you who YOU are!"** 🏘️

## What is it?

**Real-life analogy:** Imagine you move to a new school and don't know which group you belong to (Sports, Music, or Art club). You look at your **5 closest friends** (K=5). If 3 are in Sports, 1 in Music, 1 in Art → you're probably a **Sports** person!

### Main idea
1. When a **new** data point arrives, KNN measures its **distance** to ALL existing points.
2. It picks the **K nearest** neighbors.
3. **Majority vote** among those K neighbors decides the class.

### Distance formulas used:
$$\text{Euclidean Distance} = \sqrt{(x_1 - x_2)^2 + (y_1 - y_2)^2 + \dots}$$

This is just the straight-line distance between two points — like measuring with a ruler!

---

## 🔑 Key Concepts

| Concept | Simple Meaning |
|---|---|
| **K** | Number of neighbors to look at (the most important setting!) |
| **Distance** | How far apart two points are (Euclidean is default) |
| **Majority Vote** | The class with the most neighbors wins |
| **Lazy Learner** | KNN doesn't really "learn" — it just memorizes all training data! |
| **Scaling** | **CRITICAL** — because KNN uses distances, all features must be on the same scale! |

### ⚠️ How to choose K?
- **K too small** (e.g., K=1) → Noisy, overfits (one weird neighbor changes everything!)
- **K too large** (e.g., K=100) → Too smooth, underfits (ignores local patterns)
- **Sweet spot** → Usually between 3 and 15 (odd numbers preferred to avoid ties!)

# Step 4 — Data Preprocessing

### Tasks to do:
1. **Encode `species`** → numeric (setosa=0, versicolor=1, virginica=2).
   - KNN needs numbers for the target, not text!

Good news: all 4 feature columns are already numeric, so no feature encoding needed!

```python
# ----------------------------------------------------------------
# 4.1 Encode target species into numbers
# ----------------------------------------------------------------
# Label Encoding for the target (0, 1, 2 for 3 classes)
species_map = {'setosa': 0, 'versicolor': 1, 'virginica': 2}
df['species'] = df['species'].map(species_map)

print("✅ Species encoded:")
print("   setosa     -> 0")
print("   versicolor -> 1")
print("   virginica  -> 2")
print("\nEncoded target counts:")
print(df['species'].value_counts().sort_index())
```

```python
# Final check: no missing values, all numeric
print("Any missing values left?", df.isnull().sum().sum())
print("\nAll data types:")
print(df.dtypes)
```

## Step 5 — Separate Features (X) & Target (y), then Train-Test Split

- `X` = the 4 measurements (sepal & petal length/width).
- `y` = the `species` (what we want to predict).

### ⚠️ IMPORTANT for Classification: Use `stratify=y`
Ensures each species has the same ratio in both train and test sets.

```python
# ================================================================
# FEATURES (X), TARGET (y), TRAIN-TEST SPLIT
# ================================================================
X = df.drop('species', axis=1)   # 4 features
y = df['species']                 # target (0, 1, or 2)

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,       # 20% for testing (30 flowers)
    random_state=42,       # reproducible split
    stratify=y             # ✅ keep 33/33/33 ratio in both sets
)

print(f"Training flowers: {X_train.shape[0]}")
print(f"Testing flowers : {X_test.shape[0]}")
print(f"Total features  : {X_train.shape[1]}")
print("\nTrain class balance:")
print(y_train.value_counts().sort_index())
print("\nTest class balance:")
print(y_test.value_counts().sort_index())
```

# Step 6 — Feature Scaling ✅ MANDATORY for KNN!

### 💡 *Why is scaling CRITICAL for KNN?*
KNN calculates **distances** between points. Look at our features:
- `petal_length` → ranges from 1.0 to 6.9 cm
- `petal_width` → ranges from 0.1 to 2.5 cm

Without scaling, `petal_length` would dominate the distance calculation just because its numbers are bigger! That's unfair to `petal_width`.

**StandardScaler** brings every feature to $\mu = 0$, $\sigma = 1$:
$$z = \frac{x - \mu}{\sigma}$$

### ⚠️ Golden Rule: `fit_transform()` on TRAIN, only `transform()` on TEST!

```python
# ================================================================
# FEATURE SCALING (MANDATORY for KNN)
# ================================================================
scaler = StandardScaler()

# Learn mean & std from TRAINING data only, and scale it
X_train_scaled = scaler.fit_transform(X_train)

# Apply SAME scaling to TEST data
X_test_scaled = scaler.transform(X_test)

print("✅ Feature scaling done!")
print(f"Example row BEFORE scaling: {X_train.iloc[0].values}")
print(f"Example row AFTER scaling : {X_train_scaled[0].round(3)}")

# Verify: mean should now be ~0, std should be ~1
print(f"\nMean of scaled features (should be ~0): {X_train_scaled.mean():.6f}")
print(f"Std of scaled features (should be ~1) : {X_train_scaled.std():.4f}")
```

# Step 7 — Model Training

### 💡 *How KNN "learns":*
Here's the funny thing — **KNN doesn't really learn anything!** 🤯

When you call `.fit()`, KNN just **memorizes** all the training data. That's it!
All the real work happens during `.predict()`:

1. Take the new flower's measurements.
2. Calculate distance to **every** training flower.
3. Find the **K closest** flowers.
4. **Majority vote** → whatever species most of those K neighbors are, that's the prediction!

This is why KNN is called a **"Lazy Learner"** — it's lazy during training but works hard during prediction!

### Key parameters:
- **`n_neighbors=5`** → look at 5 closest flowers (default, usually a good start).
- **`weights='uniform'`** → all K neighbors get equal vote.
- **`metric='minkowski'`** → general distance formula (with p=2, it becomes Euclidean).

```python
# ================================================================
# MODEL TRAINING
# ================================================================
model = KNeighborsClassifier(
    n_neighbors=5,          # look at 5 nearest neighbors
    weights='uniform',       # all neighbors get equal vote
    metric='minkowski',      # distance formula (p=2 = Euclidean)
    p=2                      # p=2 makes it Euclidean distance
)

# .fit() just stores the training data (super fast!)
model.fit(X_train_scaled, y_train)

print("✅ KNN Model 'trained' (data memorized)!")
print(f"Number of training samples stored: {model.n_samples_fit_}")
print(f"Number of features: {model.n_features_in_}")
print(f"K (neighbors): {model.n_neighbors}")
print("\n💡 Remember: KNN is a lazy learner — all the real work happens during prediction!")
```

# Step 8 — Prediction

For each test flower, KNN finds its 5 closest training flowers and takes a majority vote.

```python
# ================================================================
# PREDICTION
# ================================================================
y_pred = model.predict(X_test_scaled)

# Create a reverse mapping to show species names
reverse_map = {0: 'setosa', 1: 'versicolor', 2: 'virginica'}

# Compare actual vs predicted for first 10 test flowers
comparison = pd.DataFrame({
    'Actual': y_test.values[:10],
    'Predicted': y_pred[:10],
    'Actual Name': [reverse_map[v] for v in y_test.values[:10]],
    'Predicted Name': [reverse_map[v] for v in y_pred[:10]]
})
comparison['Correct?'] = comparison['Actual'] == comparison['Predicted']
comparison
```

# Step 9 — Evaluate the Model

Since this is **multi-class** (3 classes), we use `average='weighted'` for Precision, Recall, and F1.

```python
# ================================================================
# MODEL EVALUATION
# ================================================================
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred, average='weighted')
rec = recall_score(y_test, y_pred, average='weighted')
f1 = f1_score(y_test, y_pred, average='weighted')

print("=" * 50)
print("   BASELINE KNN (K=5) — REPORT CARD")
print("=" * 50)
print(f"  Accuracy  : {acc:.4f} ({acc*100:.1f}%)")
print(f"  Precision : {prec:.4f}")
print(f"  Recall    : {rec:.4f}")
print(f"  F1 Score  : {f1:.4f}")
print("=" * 50)

print("\n📋 Detailed Classification Report:")
print(classification_report(y_test, y_pred, target_names=['setosa', 'versicolor', 'virginica']))
```

```python
# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(7, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['setosa', 'versicolor', 'virginica'],
            yticklabels=['setosa', 'versicolor', 'virginica'])
plt.title("Confusion Matrix")
plt.xlabel("Predicted Species")
plt.ylabel("Actual Species")
plt.show()

print("💡 The diagonal shows correct predictions.")
print("   Off-diagonal numbers show which species get confused with each other.")
```

# Step 10 — 🔍 Find the Best K (Elbow Method)

### 💡 *Why do we do this?*
The most important decision in KNN is choosing **K** (number of neighbors).

- **K=1** → Only looks at the single closest flower. Very sensitive to noise!
- **K=5** → Looks at 5 closest flowers. Usually a good balance.
- **K=50** → Looks at almost all flowers. Too smooth, loses local patterns!

The **Elbow Method** tests many K values and plots the accuracy. We pick the K where accuracy is highest (or stops improving — the "elbow").

```python
# ================================================================
# ELBOW METHOD — Find the optimal K
# ================================================================
k_range = range(1, 31)          # test K from 1 to 30
train_scores = []
test_scores = []

for k in k_range:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_scaled, y_train)
    train_scores.append(knn.score(X_train_scaled, y_train))
    test_scores.append(knn.score(X_test_scaled, y_test))

# Plot the results
plt.figure(figsize=(10, 6))
plt.plot(k_range, train_scores, 'o-', label='Training Accuracy', color='blue')
plt.plot(k_range, test_scores, 's-', label='Testing Accuracy', color='red')
plt.xlabel('K (Number of Neighbors)')
plt.ylabel('Accuracy')
plt.title('Elbow Method — Finding the Best K for KNN')
plt.xticks(k_range)
plt.legend()
plt.axvline(x=k_range[np.argmax(test_scores)], color='green', linestyle='--',
            label=f'Best K = {k_range[np.argmax(test_scores)]}')
plt.legend()
plt.show()

best_k = k_range[np.argmax(test_scores)]
best_acc = max(test_scores)
print(f"🏆 Best K = {best_k} with test accuracy = {best_acc:.4f} ({best_acc*100:.1f}%)")

print("\n💡 Observations:")
print("  - Training accuracy DECREASES as K increases (less memorization)")
print("  - Test accuracy peaks around the optimal K, then may drop")
print("  - The gap between train and test = overfitting (big gap = bad!)")
```

# Step 11 — Hyperparameter Tuning

### Key hyperparameters for KNN:
- **`n_neighbors`** → The K value (most important!).
- **`weights`** → `'uniform'` (equal vote) or `'distance'` (closer neighbors get more vote).
- **`metric`** → Distance formula (`'euclidean'`, `'manhattan'`, `'minkowski'`).

**We use `GridSearchCV` to find the best combination!**

```python
# ================================================================
# HYPERPARAMETER TUNING with GridSearchCV
# ================================================================
param_grid = {
    'n_neighbors': list(range(1, 31, 2)),   # odd numbers only: 1, 3, 5, ..., 29
    'weights': ['uniform', 'distance'],
    'metric': ['euclidean', 'manhattan', 'minkowski']
}

grid = GridSearchCV(
    KNeighborsClassifier(),
    param_grid,
    cv=5,                # 5-fold cross-validation
    scoring='accuracy',   # optimize accuracy (Iris is balanced, so accuracy is fine)
    n_jobs=-1,
    verbose=1
)

print("⏳ Tuning is running...")
grid.fit(X_train_scaled, y_train)

print("\n✅ Best hyperparameters found:")
print(grid.best_params_)
print(f"Best cross-validation accuracy: {grid.best_score_:.4f}")

# Use the best model
best_model = grid.best_estimator_
y_pred_tuned = best_model.predict(X_test_scaled)

print("\n--- Tuned Model on Test Data ---")
print(f"Accuracy  : {accuracy_score(y_test, y_pred_tuned):.4f}")
print(f"Precision : {precision_score(y_test, y_pred_tuned, average='weighted'):.4f}")
print(f"Recall    : {recall_score(y_test, y_pred_tuned, average='weighted'):.4f}")
print(f"F1 Score  : {f1_score(y_test, y_pred_tuned, average='weighted'):.4f}")
```

```python
# ================================================================
# FINAL COMPARISON: Baseline vs Tuned KNN
# ================================================================
comparison_df = pd.DataFrame({
    'Model': ['Baseline KNN (K=5)', f'Tuned KNN (K={best_model.n_neighbors})'],
    'Accuracy':  [accuracy_score(y_test, y_pred),   accuracy_score(y_test, y_pred_tuned)],
    'Precision': [precision_score(y_test, y_pred, average='weighted'),
                  precision_score(y_test, y_pred_tuned, average='weighted')],
    'Recall':    [recall_score(y_test, y_pred, average='weighted'),
                  recall_score(y_test, y_pred_tuned, average='weighted')],
    'F1 Score':  [f1_score(y_test, y_pred, average='weighted'),
                  f1_score(y_test, y_pred_tuned, average='weighted')]
})

print("=" * 70)
print("FINAL MODEL COMPARISON")
print("=" * 70)
comparison_df.round(4)
```

```python
# Tuned model confusion matrix
cm_tuned = confusion_matrix(y_test, y_pred_tuned)

plt.figure(figsize=(7, 5))
sns.heatmap(cm_tuned, annot=True, fmt='d', cmap='Greens',
            xticklabels=['setosa', 'versicolor', 'virginica'],
            yticklabels=['setosa', 'versicolor', 'virginica'])
plt.title(f"Confusion Matrix — Tuned KNN (K={best_model.n_neighbors})")
plt.xlabel("Predicted Species")
plt.ylabel("Actual Species")
plt.show()
```

Save both the trained model **and** the scaler — we need the scaler to transform any new flower's measurements before predicting!

### 🏆 Golden Rules for KNN
1. **ALWAYS scale features** — KNN uses distances, so scaling is CRITICAL!
2. **Always use `stratify=y`** for classification splits.
3. **Use ODD K values** (3, 5, 7...) to avoid ties in voting.
4. **Small K = overfitting, Large K = underfitting** — find the sweet spot!
5. **KNN is a lazy learner** — fast to train, slow to predict on big data.
6. **Always save the scaler** — you need it to predict new samples correctly!
7. **Elbow Method** is the best visual way to choose K.

### 🌟 Pros & Cons of KNN

| ✅ Pros | ❌ Cons |
|---|---|
| Super simple to understand | Slow prediction on large datasets |
| No training time (lazy learner) | Uses lots of memory (stores ALL data) |
| Works for both classification & regression | Requires scaling (extra step) |
| Naturally handles multi-class problems | Sensitive to irrelevant features |
| No assumptions about data distribution | Curse of dimensionality (bad with many features) |
| Adapts easily to new training data | Choosing the right K takes effort |

### 🎯 Key Concepts to Remember
1. **K** = number of neighbors to vote (the most important setting!).
2. **Distance** = how we measure "closeness" (Euclidean is default).
3. **Majority Vote** = the class with the most neighbors wins.
4. **Lazy Learner** = KNN memorizes data instead of learning a formula.
5. **Elbow Method** = plot accuracy vs K to find the best K.
6. **Scaling** = MANDATORY because KNN uses a ruler to measure distances!
