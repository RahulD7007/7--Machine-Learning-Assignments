### What is a Decision Tree?
A Decision Tree is like playing the game **'20 Questions'** with a computer!
It asks a series of Yes/No questions to reach a final decision.

## What is it?

A Decision Tree makes predictions by asking simple questions about the features, one at a time.

Example:

**Is age < 10?** → Yes → **Is sex = female?** → Yes → 🟢 Survived

### Main idea
The tree finds the **best question** at each step that splits the data into the purest possible groups (mostly survivors OR mostly non-survivors).

It uses mathematical measures like:
- **Gini Impurity** (default) — measures how mixed a group is.
- **Entropy** — measures the disorder/randomness in a group.

$$\text{Gini} = 1 - \sum_{i=1}^{n} p_i^2$$

- Gini = 0 → Pure group (all same class) ✅
- Gini = 0.5 → Perfectly mixed group ❌

# Step 4 — Data Preprocessing

### Tasks to do:
1. **Drop useless columns** — `PassengerId`, `Name`, `Ticket`, `Cabin`.
   - `Cabin` has too many missing values (over 70%).
   - `Name`, `Ticket`, `PassengerId` are unique IDs → not useful patterns.
2. **Fill missing `Age`** with the median.
3. **Fill missing `Embarked`** with the mode (most common value).
4. **Encode text columns** (`Sex`, `Embarked`) into numbers — computers only understand numbers!

```python
# ----------------------------------------------------------------
# 4.1 Drop useless columns
# ----------------------------------------------------------------
cols_to_drop = ['PassengerId', 'Name', 'Ticket', 'Cabin']
df = df.drop(columns=cols_to_drop)

print("✅ Dropped columns:", cols_to_drop)
print("Remaining columns:", df.columns.tolist())
```

```python
# ----------------------------------------------------------------
# 4.2 Fill missing values
# ----------------------------------------------------------------
# Age → fill with median (robust against outliers)
df['Age'] = df['Age'].fillna(df['Age'].median())

# Embarked → fill with mode (most common port)
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

# Fare (sometimes has 1 missing value) → fill with median
df['Fare'] = df['Fare'].fillna(df['Fare'].median())

print("✅ Missing values fixed!")
print("Missing values now:", df.isnull().sum().sum())
```

```python
# ----------------------------------------------------------------
# 4.3 Encode text columns into numbers
# ----------------------------------------------------------------
# Sex → male=0, female=1 (Label Encoding since only 2 categories)
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})

# Embarked → One-Hot Encoding (3 categories: C, Q, S)
df = pd.get_dummies(df, columns=['Embarked'], drop_first=True)

# Convert True/False from get_dummies to int (0/1) for cleaner display
bool_cols = df.select_dtypes(include='bool').columns
df[bool_cols] = df[bool_cols].astype(int)

print("✅ Encoding complete!")
df.head()
```

```python
# Final check: no missing values and all numeric columns
print("Any missing values left?", df.isnull().sum().sum())
print("\nAll data types now:")
print(df.dtypes)
```

```python
# Now that everything is numeric, let's see the correlation heatmap
plt.figure(figsize=(9, 7))
sns.heatmap(df.corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title("Feature Correlation Heatmap")
plt.show()

print("Correlation with Survived (sorted):")
print(df.corr()['Survived'].sort_values(ascending=False))
```

## Step 5 — Separate Features (X) & Target (y), then Train-Test Split

- `X` = all the **clues** (input columns).
- `y` = the **answer** (`Survived`) we want to predict.

### ⚠️ IMPORTANT for Classification: Use `stratify=y`
Keeps class balance the same in both train and test sets.

```python
# ================================================================
# FEATURES (X), TARGET (y), TRAIN-TEST SPLIT
# ================================================================
X = df.drop('Survived', axis=1)   # everything EXCEPT Survived
y = df['Survived']                 # what we want to predict

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,       # 20% for testing
    random_state=42,       # reproducible split
    stratify=y             # ✅ keep class balance in both sets
)

print(f"Training passengers: {X_train.shape[0]}")
print(f"Testing passengers : {X_test.shape[0]}")
print("\nTrain class balance:")
print(y_train.value_counts(normalize=True).round(2))
print("\nTest class balance:")
print(y_test.value_counts(normalize=True).round(2))
```

# Step 6 — Feature Scaling ❌ NOT NEEDED!

### 💡 *Why don't we scale for Decision Trees?*
Decision Trees make decisions like:
> "Is `Age < 10`?" or "Is `Fare > 50`?"

The **actual scale doesn't matter** — the tree just asks whether a value is above or below a threshold. Whether age is in years (0-80) or seconds (0-2.5 billion), the *ordering* of values is the same, and the split rule remains valid!

> **Save yourself time — skip scaling for tree-based models!**

# Step 7 — Model Training

### 💡 *How Decision Tree learns:*
1. Start with all training data at the top (**root node**).
2. Try every possible question ("Is feature X > threshold?") on every feature.
3. Pick the question that creates the **purest split** (lowest Gini impurity).
4. Repeat for each new branch until:
   - The group is pure (all same class), OR
   - `max_depth` is reached, OR
   - Too few samples to split further.

### Key parameters:
- `random_state=42` → reproducible results
- `max_depth=None` (default) → tree grows until pure (risk of overfitting!)

```python
# ================================================================
# MODEL TRAINING
# ================================================================
model = DecisionTreeClassifier(random_state=42)

# .fit() is where the actual learning happens
model.fit(X_train, y_train)

print("✅ Model trained!")
print(f"Tree depth grown: {model.get_depth()}")
print(f"Number of leaves: {model.get_n_leaves()}")

# Feature importance: which features did the tree use most?
importance_df = pd.DataFrame({
    'Feature': X.columns,
    'Importance': model.feature_importances_
}).sort_values(by='Importance', ascending=False)

print("\nFeature Importance (which clues mattered most):")
print(importance_df.round(3))
```

```python
# Visualize feature importance as a bar chart
plt.figure(figsize=(8, 5))
sns.barplot(data=importance_df, x='Importance', y='Feature', palette='viridis')
plt.title("Which Features Did the Tree Use Most?")
plt.show()
```

# Step 8 — Prediction

The tree traces each test passenger through its questions until reaching a leaf → outputs the class.

```python
# ================================================================
# PREDICTION
# ================================================================
y_pred = model.predict(X_test)

# Compare actual vs predicted for first 10 test passengers
comparison = pd.DataFrame({
    'Actual': y_test.values[:10],
    'Predicted': y_pred[:10]
})
comparison['Correct?'] = comparison['Actual'] == comparison['Predicted']
comparison
```

# Step 9 — Evaluate the Model

For classification, we use multiple metrics:

| Metric | Meaning |
|---|---|
| **Accuracy** | Overall correctness |
| **Precision** | Of predicted survivors, how many actually survived? |
| **Recall** | Of actual survivors, how many did we catch? |
| **F1 Score** | Balance of Precision & Recall |
| **Confusion Matrix** | Full breakdown of correct/wrong predictions |

```python
# ================================================================
# MODEL EVALUATION
# ================================================================
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("=" * 50)
print("   BASELINE DECISION TREE — REPORT CARD")
print("=" * 50)
print(f"  Accuracy  : {acc:.4f} ({acc*100:.1f}%)")
print(f"  Precision : {prec:.4f}")
print(f"  Recall    : {rec:.4f}")
print(f"  F1 Score  : {f1:.4f}")
print("=" * 50)

print("\n📋 Detailed Classification Report:")
print(classification_report(y_test, y_pred, target_names=['Died (0)', 'Survived (1)']))
```

```python
# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Pred Died', 'Pred Survived'],
            yticklabels=['Actual Died', 'Actual Survived'])
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

tn, fp, fn, tp = cm.ravel()
print(f"✅ True Negatives  (TN): {tn}  -> Correctly predicted 'Died'")
print(f"❌ False Positives (FP): {fp}  -> Predicted 'Survived' but actually died")
print(f"❌ False Negatives (FN): {fn}  -> Predicted 'Died' but actually survived")
print(f"✅ True Positives  (TP): {tp}  -> Correctly predicted 'Survived'")
```

# Step 10 — Hyperparameter Tuning

### ⚠️ The BIG problem with Decision Trees: **Overfitting!**
By default, a tree grows until every leaf is pure — this means it **memorizes** the training data and performs poorly on new data.

### Key hyperparameters to control tree size:
- **`max_depth`** → maximum depth of the tree (smaller = simpler, less overfitting).
- **`min_samples_split`** → minimum samples needed to split a node.
- **`min_samples_leaf`** → minimum samples required at a leaf.
- **`criterion`** → `'gini'` or `'entropy'` (both usually work similarly).

**We use `GridSearchCV` to find the best combination automatically!**

```python
# ================================================================
# HYPERPARAMETER TUNING with GridSearchCV
# ================================================================
param_grid = {
    'max_depth': [3, 5, 7, 10, None],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4],
    'criterion': ['gini', 'entropy']
}

grid = GridSearchCV(
    DecisionTreeClassifier(random_state=42),
    param_grid,
    cv=5,                # 5-fold cross-validation
    scoring='f1',         # optimize F1 (balance of precision & recall)
    n_jobs=-1             # use all CPU cores
)

grid.fit(X_train, y_train)

print("✅ Best hyperparameters found:")
print(grid.best_params_)
print(f"Best cross-validation F1 score: {grid.best_score_:.4f}")

# Use the best model
best_model = grid.best_estimator_
y_pred_tuned = best_model.predict(X_test)

print("\n--- Tuned Model on Test Data ---")
print(f"Accuracy  : {accuracy_score(y_test, y_pred_tuned):.4f}")
print(f"Precision : {precision_score(y_test, y_pred_tuned):.4f}")
print(f"Recall    : {recall_score(y_test, y_pred_tuned):.4f}")
print(f"F1 Score  : {f1_score(y_test, y_pred_tuned):.4f}")
print(f"Tree depth: {best_model.get_depth()}")
print(f"Leaves    : {best_model.get_n_leaves()}")
```

```python
# ================================================================
# FINAL COMPARISON: Baseline vs Tuned Tree
# ================================================================
comparison_df = pd.DataFrame({
    'Model': ['Baseline Decision Tree', 'Tuned Decision Tree'],
    'Accuracy':  [accuracy_score(y_test, y_pred),   accuracy_score(y_test, y_pred_tuned)],
    'Precision': [precision_score(y_test, y_pred),  precision_score(y_test, y_pred_tuned)],
    'Recall':    [recall_score(y_test, y_pred),     recall_score(y_test, y_pred_tuned)],
    'F1 Score':  [f1_score(y_test, y_pred),         f1_score(y_test, y_pred_tuned)]
})

print("=" * 70)
print("FINAL MODEL COMPARISON")
print("=" * 70)
comparison_df.round(4)
```

## Step 11 — 🎨 Visualize the Decision Tree!

This is the **coolest part of Decision Trees** — you can literally SEE how the model makes decisions!

Each box shows:
- The **question** being asked (e.g., `Sex <= 0.5`)
- The **gini** impurity
- **samples**: how many training passengers reached this point
- **value**: [# who died, # who survived] in this group

```python
# ================================================================
# VISUALIZE THE TREE (using the tuned model)
# ================================================================
plt.figure(figsize=(20, 10))
plot_tree(best_model,
          feature_names=X.columns,
          class_names=['Died', 'Survived'],
          filled=True,             # color the boxes
          rounded=True,             # rounded box corners
          fontsize=10)
plt.title(f"Decision Tree (depth={best_model.get_depth()}) — How the Model Decides Who Survives", fontsize=14)
plt.show()
```

```python
# Bonus: text version of the tree (great for reading through the rules)
tree_rules = export_text(best_model, feature_names=list(X.columns))
print("Decision rules the tree learned:\n")
print(tree_rules)
```

## Step 12 — Save the Final Model

Save the trained model so we can reuse it later without retraining.

```python
# ================================================================
# SAVE THE FINAL MODEL
# ================================================================
final_model = best_model   # use the tuned model

joblib.dump(final_model, "decision_tree_titanic_model.pkl")

# Save the list of feature names too (very useful for predicting new samples)
joblib.dump(list(X.columns), "titanic_feature_names.pkl")

print("✅ Model saved as 'decision_tree_titanic_model.pkl'")
print("✅ Feature names saved as 'titanic_feature_names.pkl'")
```

```python
# ================================================================
# BONUS: Load the saved model and predict for a NEW passenger
# ================================================================
loaded_model = joblib.load("decision_tree_titanic_model.pkl")
feature_names = joblib.load("titanic_feature_names.pkl")

# Example: young female, 1st class, no family
new_passenger = pd.DataFrame([{
    'Pclass': 1,
    'Sex': 1,            # 1 = female, 0 = male
    'Age': 22,
    'SibSp': 0,
    'Parch': 0,
    'Fare': 80.0,
    'Embarked_Q': 0,
    'Embarked_S': 1
}])[feature_names]   # ensure same column order

predicted_class = loaded_model.predict(new_passenger)[0]
predicted_proba = loaded_model.predict_proba(new_passenger)[0][1]

print("=" * 45)
print("   NEW PASSENGER SURVIVAL PREDICTION")
print("=" * 45)
print(f"Prediction: {'🟢 Survived' if predicted_class == 1 else '🔴 Died'}")
print(f"Survival probability: {predicted_proba*100:.1f}%")
print("=" * 45)
```

| **12. Visualize** | `plot_tree()` | Literally SEE how the model decides! |
| **13. Save** | `joblib.dump()` | Store our trained brain for later use. |

### 🏆 Golden Rules for Decision Trees
1. **NO scaling needed** — trees split on raw values.
2. **Always use `stratify=y`** for classification splits.
3. **Watch out for overfitting** — always tune `max_depth`!
4. **Feature importance** tells you which clues the tree found most useful.
5. **Trees are interpretable** — you can visualize & explain every decision.
6. **Encode categorical text** (Sex, Embarked) into numbers before training.

### 🌟 Pros & Cons of Decision Trees

| ✅ Pros | ❌ Cons |
|---|---|
| Easy to understand & visualize | Prone to overfitting |
| No scaling required | Unstable (small data change → different tree) |
| Handles both numeric & categorical | Can be biased with imbalanced data |
| Fast to train | Not always the most accurate |
| Shows feature importance | Better versions exist (Random Forest, XGBoost) |
