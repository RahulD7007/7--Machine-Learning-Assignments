### What is Logistic Regression?
Despite having the word "Regression" in its name, Logistic Regression is actually a **CLASSIFICATION** algorithm!
It predicts a **category** (like Yes/No, 0/1, Diabetic/Not Diabetic) instead of a continuous number.

## What is it?

**Real-life analogy:** Imagine a doctor looking at a patient's blood sugar, BMI, and age. Based on these numbers, the doctor estimates the **probability** (chance) that the patient has diabetes — say 78%. If it's above 50%, the doctor says "Yes, diabetic."

That's exactly what Logistic Regression does!

### Main idea
Instead of drawing a straight line through the data (which could output weird values like -5 or 200%), it fits an **S-shaped curve** called the **Sigmoid function**:

$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

- If output probability $\ge 0.5$ $\rightarrow$ Class **1** (Diabetic)
- If output probability $< 0.5$ $\rightarrow$ Class **0** (Not Diabetic)

---

## 🔑 Linear vs Logistic Regression — Key Differences

| Feature | Linear Regression | Logistic Regression |
|---|---|---|
| **Output** | Continuous number (e.g. $250,000) | Probability (0 to 1) $\rightarrow$ Class (0 or 1) |
| **Task Type** | Regression | **Classification** |
| **Fit Shape** | Straight line / plane | S-shaped curve (Sigmoid) |
| **Key Metrics** | MAE, RMSE, R² | Accuracy, Precision, Recall, F1, ROC-AUC |
| **Split Rule** | `stratify=None` | **`stratify=y` (crucial to keep class ratio)** |

# Step 4 — Data Preprocessing

### 🚨 Medical Anomaly: Hidden Missing Values (Impossible Zeros)
In this dataset, values of **0** for `Glucose`, `BloodPressure`, `SkinThickness`, `Insulin`, and `BMI` are **biologically impossible** for a living person.

These are **missing values recorded as 0** — a common data entry shortcut.

**Strategy:**
1. Replace impossible zeros with `np.nan`.
2. Impute `NaN` values with the column **median** (robust against outliers).

```python
# ----------------------------------------------------------------
# Fix the hidden missing values — impossible 0s in medical columns
# ----------------------------------------------------------------
invalid_zero_cols = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

print("Count of impossible 0s BEFORE cleaning:")
for col in invalid_zero_cols:
    zero_count = (df[col] == 0).sum()
    print(f"  {col:20s}: {zero_count:3d} zeros ({zero_count/len(df)*100:.1f}%)")

# Step 1: Replace 0 with NaN
df[invalid_zero_cols] = df[invalid_zero_cols].replace(0, np.nan)

# Step 2: Impute NaN with median values
for col in invalid_zero_cols:
    median_val = df[col].median()
    df[col] = df[col].fillna(median_val)

print("\n✅ Preprocessing complete! Remaining NaNs:", df.isnull().sum().sum())
```

```python
# Verify: no missing values, all numeric
print("Any missing values left?", df.isnull().sum().sum())
print("\nAll data types:")
print(df.dtypes)
```

## Step 5 — Separate Features (X) & Target (y), then Train-Test Split

- `X` = all 8 features (the clues).
- `y` = the `Outcome` (what we want to predict).

### ⚠️ IMPORTANT for Classification: Use `stratify=y`
`stratify=y` ensures both training and testing sets have the **same ratio** of Diabetic (~35%) to Non-Diabetic (~65%) patients. Without it, one set might accidentally get too many of one class!

```python
# ================================================================
# FEATURES (X), TARGET (y), TRAIN-TEST SPLIT
# ================================================================
X = df.drop('Outcome', axis=1)   # 8 features
y = df['Outcome']                 # target (0 or 1)

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,       # 20% held out for testing
    random_state=42,       # reproducible split
    stratify=y             # ✅ preserve class distribution in both splits
)

print(f"Training patients: {X_train.shape[0]}")
print(f"Testing patients : {X_test.shape[0]}")
print(f"Total features   : {X_train.shape[1]}")
print("\nTrain class balance:")
print(y_train.value_counts(normalize=True).round(2))
print("\nTest class balance:")
print(y_test.value_counts(normalize=True).round(2))
```

# Step 6 — Feature Scaling ✅ REQUIRED for Logistic Regression!

### 💡 *Why do we scale?*
Logistic Regression uses **gradient descent** to find the best weights. If features have very different ranges:
- `Glucose` → 44 to 199
- `DiabetesPedigreeFunction` → 0.08 to 2.42

The optimizer will struggle because `Glucose` dominates the gradient. Scaling makes all features equally important.

**StandardScaler** brings every feature to $\mu = 0$, $\sigma = 1$:
$$z = \frac{x - \mu}{\sigma}$$

### ⚠️ Golden Rule: `fit_transform()` on TRAIN, only `transform()` on TEST!
- 🛑 **Never fit on test data!** That's called **data leakage**.

```python
# ================================================================
# FEATURE SCALING (REQUIRED for Logistic Regression)
# ================================================================
scaler = StandardScaler()

# Learn mean & std from TRAINING data only, then scale it
X_train_scaled = scaler.fit_transform(X_train)

# Apply SAME scaling to TEST data (never re-learn from test!)
X_test_scaled = scaler.transform(X_test)

print("✅ Feature scaling done!")
print(f"Example row BEFORE scaling: {X_train.iloc[0].values[:4].round(2)}")
print(f"Example row AFTER scaling : {X_train_scaled[0][:4].round(2)}")

# Verify scaling worked
print(f"\nMean of scaled train (should be ~0): {X_train_scaled.mean():.6f}")
print(f"Std of scaled train (should be ~1) : {X_train_scaled.std():.4f}")
```

# Step 7 — Model Training

### 💡 *How Logistic Regression learns:*
1. Computes a weighted sum of all features:
   $$z = w_1 x_1 + w_2 x_2 + \dots + w_n x_n + b$$
2. Passes $z$ through the **Sigmoid function** to get a probability:
   $$P(y=1) = \frac{1}{1 + e^{-z}}$$
3. Adjusts weights using **gradient descent** to minimize the **log loss**.

### Key parameters:
- **`max_iter=1000`** → gives the optimizer enough steps to converge.
- **`random_state=42`** → reproducibility.
- **`C=1.0`** (default) → inverse of regularization strength.

```python
# ================================================================
# MODEL TRAINING
# ================================================================
model = LogisticRegression(max_iter=1000, random_state=42)

# .fit() learns the optimal weights and bias
model.fit(X_train_scaled, y_train)

print("✅ Logistic Regression model trained!")
print(f"Intercept (b): {model.intercept_[0]:.3f}")
print("\nLearned Feature Coefficients (Weights):")

coef_df = pd.DataFrame({
    'Feature': X.columns,
    'Weight': model.coef_[0]
}).sort_values(by='Weight', key=abs, ascending=False)

for _, row in coef_df.iterrows():
    direction = "↑ Increases Diabetes Risk" if row['Weight'] > 0 else "↓ Decreases Diabetes Risk"
    print(f"  {row['Feature']:28s} -> {row['Weight']:>7.3f}  ({direction})")
```

# Step 8 — Prediction

Two useful outputs:
- `.predict()` → final class (0 or 1) using 0.5 threshold.
- `.predict_proba()` → probability of each class (very useful in medical decisions!).

```python
# ================================================================
# PREDICTION
# ================================================================
y_pred = model.predict(X_test_scaled)                # 0 or 1
y_proba = model.predict_proba(X_test_scaled)[:, 1]   # probability of class 1 (Diabetic)

# Compare actual vs predicted for first 10 test patients
comparison = pd.DataFrame({
    'Actual': y_test.values[:10],
    'Predicted': y_pred[:10],
    'Diabetes Probability': y_proba[:10].round(3)
})
comparison['Correct?'] = comparison['Actual'] == comparison['Predicted']
comparison
```

# Step 9 — Evaluate the Model

| Metric | Meaning |
|---|---|
| **Accuracy** | Overall correctness |
| **Precision** | Of predicted diabetics, how many actually had it? |
| **Recall (Sensitivity)** | Of actual diabetics, how many did we catch? |
| **F1 Score** | Balance of Precision & Recall |
| **ROC-AUC** | Overall separator quality (0.5 = random, 1.0 = perfect) |

### ⚠️ In medical data, high **RECALL** is CRITICAL!
Missing a diabetic patient (**False Negative**) can be life-threatening! A false alarm (**False Positive**) just means an extra test — much less dangerous.

```python
# ================================================================
# MODEL EVALUATION
# ================================================================
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
auc = roc_auc_score(y_test, y_proba)

print("=" * 55)
print("   BASELINE LOGISTIC REGRESSION — REPORT CARD")
print("=" * 55)
print(f"  Accuracy  : {acc:.4f} ({acc*100:.1f}%)")
print(f"  Precision : {prec:.4f}")
print(f"  Recall    : {rec:.4f}")
print(f"  F1 Score  : {f1:.4f}")
print(f"  ROC-AUC   : {auc:.4f}")
print("=" * 55)

print("\n📋 Detailed Classification Report:")
print(classification_report(y_test, y_pred, target_names=['No Diabetes (0)', 'Diabetes (1)']))
```

```python
# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Pred No Diabetes', 'Pred Diabetes'],
            yticklabels=['Actual No Diabetes', 'Actual Diabetes'])
plt.title("Confusion Matrix")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()

tn, fp, fn, tp = cm.ravel()
print(f"✅ True Negatives  (TN): {tn}  -> Correctly said 'No Diabetes'")
print(f"❌ False Positives (FP): {fp}  -> Said 'Diabetes' but was healthy (false alarm)")
print(f"⚠️ False Negatives (FN): {fn}  -> MISSED a diabetes case (DANGEROUS!)")
print(f"✅ True Positives  (TP): {tp}  -> Correctly detected diabetes")
```

```python
# ROC Curve
fpr, tpr, thresholds = roc_curve(y_test, y_proba)

plt.figure(figsize=(7, 6))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'Logistic Regression (AUC = {auc:.3f})')
plt.plot([0, 1], [0, 1], color='gray', lw=1, linestyle='--', label='Random guess (AUC = 0.5)')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate (Recall)')
plt.title('ROC Curve — How well does our model separate classes?')
plt.legend(loc='lower right')
plt.show()
```

# Step 10 — Hyperparameter Tuning

### Key hyperparameters for Logistic Regression:
- **`C`** → Inverse of regularization strength. Smaller C = stronger penalty (simpler model).
- **`penalty`** → `'l1'` (Lasso, can zero-out features) or `'l2'` (Ridge, shrinks features).
- **`solver`** → `'liblinear'` supports both L1 and L2 penalties.

**We use `GridSearchCV` to find the best combination!**

```python
# ================================================================
# HYPERPARAMETER TUNING with GridSearchCV
# ================================================================
param_grid = {
    'C': [0.001, 0.01, 0.1, 1.0, 10.0, 100.0],
    'penalty': ['l1', 'l2'],
    'solver': ['liblinear']
}

grid = GridSearchCV(
    LogisticRegression(max_iter=1000, random_state=42),
    param_grid,
    cv=5,                # 5-fold cross-validation
    scoring='f1',         # optimize F1 (good for medical data)
    n_jobs=-1
)

grid.fit(X_train_scaled, y_train)

print("✅ Best hyperparameters found:")
print(grid.best_params_)
print(f"Best cross-validation F1 score: {grid.best_score_:.4f}")

# Use the best model
best_model = grid.best_estimator_
y_pred_tuned = best_model.predict(X_test_scaled)
y_proba_tuned = best_model.predict_proba(X_test_scaled)[:, 1]

print("\n--- Tuned Model on Test Data ---")
print(f"Accuracy  : {accuracy_score(y_test, y_pred_tuned):.4f}")
print(f"Precision : {precision_score(y_test, y_pred_tuned):.4f}")
print(f"Recall    : {recall_score(y_test, y_pred_tuned):.4f}")
print(f"F1 Score  : {f1_score(y_test, y_pred_tuned):.4f}")
print(f"ROC-AUC   : {roc_auc_score(y_test, y_proba_tuned):.4f}")
```

```python
# ================================================================
# FINAL COMPARISON: Baseline vs Tuned
# ================================================================
comparison_df = pd.DataFrame({
    'Model': ['Baseline Logistic Regression', 'Tuned Logistic Regression'],
    'Accuracy':  [accuracy_score(y_test, y_pred),   accuracy_score(y_test, y_pred_tuned)],
    'Precision': [precision_score(y_test, y_pred),  precision_score(y_test, y_pred_tuned)],
    'Recall':    [recall_score(y_test, y_pred),     recall_score(y_test, y_pred_tuned)],
    'F1 Score':  [f1_score(y_test, y_pred),         f1_score(y_test, y_pred_tuned)],
    'ROC-AUC':   [roc_auc_score(y_test, y_proba),   roc_auc_score(y_test, y_proba_tuned)]
})

print("=" * 75)
print("FINAL MODEL COMPARISON")
print("=" * 75)
comparison_df.round(4)
```

Save both the trained model **and** the scaler — we need the scaler to transform any new patient data before predicting!

```python
# ================================================================
# BONUS: Load the saved model and predict for a NEW patient
# ================================================================
loaded_model = joblib.load("logistic_regression_diabetes_model.pkl")
loaded_scaler = joblib.load("diabetes_scaler.pkl")
feature_names = joblib.load("diabetes_features.pkl")

# Build a new patient record
new_patient = pd.DataFrame([{
    'Pregnancies': 3,
    'Glucose': 145.0,
    'BloodPressure': 78.0,
    'SkinThickness': 28.0,
    'Insulin': 130.0,
    'BMI': 32.4,
    'DiabetesPedigreeFunction': 0.65,
    'Age': 42
}])[feature_names]   # ensure exact column order

# IMPORTANT: Scale using the SAME scaler from training!
new_patient_scaled = loaded_scaler.transform(new_patient)

# Predict
pred_class = loaded_model.predict(new_patient_scaled)[0]
pred_proba = loaded_model.predict_proba(new_patient_scaled)[0][1]

print("=" * 50)
print("   NEW PATIENT DIABETES PREDICTION")
print("=" * 50)
print(f"Prediction: {'🔴 Diabetic' if pred_class == 1 else '🟢 Not Diabetic'}")
print(f"Diabetes probability: {pred_proba*100:.1f}%")
print("=" * 50)

if pred_proba > 0.5:
    print("\n💡 Medical Action: Recommend further glucose testing & lifestyle changes!")
else:
    print("\n💡 Medical Action: Continue regular health check-ups.")
```

### 🏆 Golden Rules for Logistic Regression
1. **ALWAYS scale features** — Logistic Regression uses gradient descent.
2. **Always use `stratify=y`** for classification splits.
3. **In medical data, check for impossible zeros** — they're hidden missing values.
4. **Accuracy alone is misleading** — always check Recall, Precision, and F1.
5. **High Recall is essential in healthcare** — a False Negative (missing a sick patient) is dangerous.
6. **Always save the scaler** — you need it to predict new samples correctly!
7. **`max_iter=1000`** prevents convergence warnings.

### 🌟 Pros & Cons of Logistic Regression

| ✅ Pros | ❌ Cons |
|---|---|
| Simple, fast, and interpretable | Assumes linear relationship between features and log-odds |
| Gives probability outputs | Struggles with complex non-linear patterns |
| Works well on small datasets | Sensitive to outliers |
| No overfitting with regularization | Requires scaling |
| Great baseline for classification | Can't handle multi-class without modifications |
| Coefficients show feature importance | Assumes no multicollinearity |
