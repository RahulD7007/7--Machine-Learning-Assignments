### What is Logistic Regression?
Despite having the word "Regression" in its name, Logistic Regression is actually a **CLASSIFICATION** algorithm!
It predicts a **category/class** (like Yes/No, 0/1, Diabetic/Not Diabetic) instead of a continuous number.

## What is it?

Logistic Regression predicts a **probability between 0 and 1**, then converts it into a class label.

Example:

**Patient Health Clues → Model → Probability (e.g. 0.82) → 1 (Diabetic)**

### Main idea
Instead of drawing a straight line through the data (which could output weird values like negative numbers or numbers greater than 100%), it fits an **S-shaped curve** called the **Sigmoid function**:

$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

- If output probability $\ge 0.5$ $\rightarrow$ Class **1** (Diabetic)
- If output probability $< 0.5$ $\rightarrow$ Class **0** (Not Diabetic)

---

## 🔑 Linear vs Logistic Regression — Key Differences

| Feature | Linear Regression | Logistic Regression |
|---|---|---|
| **Output** | Continuous number (e.g. $250,000) | Probability (0 to 1) $\rightarrow$ Class (0 or 1) |
| **Task Type** | Regression | **Classification** |
| **Fit Shape** | Straight line / plane | S-shaped curve (Sigmoid) |
| **Key Metrics** | MAE, RMSE, R² | Accuracy, Precision, Recall, F1, ROC-AUC |
| **Split Rule** | `stratify=None` | **`stratify=y` (crucial to keep class ratio)** |

# Step 4 — Data Preprocessing & Cleaning

### 🚨 Medical Anomaly: Hidden Missing Values (Zero Values)
In physiological measurements, values of **0** for features such as **Glucose, Blood Pressure, Skin Thickness, Insulin, and BMI** are biologically impossible for a living patient.

These are **missing values recorded as 0**.

**Strategy:**
1. Replace impossible zeros with `np.nan`.
2. Impute `NaN` values with the column **median** (robust against outliers).

```python
# Columns that cannot biologically have a value of 0
invalid_zero_cols = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

print("Count of impossible 0s BEFORE cleaning:")
for col in invalid_zero_cols:
    zero_count = (df[col] == 0).sum()
    print(f"  {col:25s}: {zero_count:3d} zeros ({zero_count/len(df)*100:.1f}%)")

# Step 1: Replace 0 with NaN
df[invalid_zero_cols] = df[invalid_zero_cols].replace(0, np.nan)

# Step 2: Impute NaN with median values
for col in invalid_zero_cols:
    median_val = df[col].median()
    df[col] = df[col].fillna(median_val)

print("\n✅ Preprocessing Complete! Remaining NaNs:", df.isnull().sum().sum())
```

## Step 5 — Separate Features (X) & Target (y), then Train-Test Split

- `X` = Input features/clues (all columns except `Outcome`).
- `y` = The target category to predict (`Outcome`).

### ⚠️ Golden Rule for Classification: Use `stratify=y`
`stratify=y` guarantees that both the training set (80%) and test set (20%) have the **exact same ratio** of Diabetic (35%) to Non-Diabetic (65%) patients, preventing sampling bias.

```python
# ================================================================
# FEATURES (X), TARGET (y), TRAIN-TEST SPLIT
# ================================================================
X = df.drop('Outcome', axis=1)
y = df['Outcome']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,      # 20% held out for evaluation
    random_state=42,      # deterministic reproducibility
    stratify=y            # preserves class distribution in both splits
)

print(f"Training Set  : {X_train.shape[0]} samples")
print(f"Testing Set   : {X_test.shape[0]} samples")
print("\nTraining Class Ratio:")
print(y_train.value_counts(normalize=True).round(3))
print("\nTesting Class Ratio:")
print(y_test.value_counts(normalize=True).round(3))
```

# Step 6 — Feature Scaling

### 💡 *Why do we scale for Logistic Regression?*
Features have very different units and magnitudes (e.g. `Age` $\in [21, 81]$, `Insulin` $\in [14, 846]$, `DiabetesPedigreeFunction` $\in [0.07, 2.42]$).

**StandardScaler** standardizes all features such that:
$$z = \frac{x - \mu}{\sigma} \quad (\mu = 0, \; \sigma = 1)$$

### ⚠️ Golden Rule: `fit_transform()` on Train, only `transform()` on Test!
- Fit the mean $\mu$ and standard deviation $\sigma$ **only** on `X_train`.
- Apply those exact parameters to transform `X_test`.
- This strictly prevents **Data Leakage**.

```python
# ================================================================
# FEATURE SCALING
# ================================================================
scaler = StandardScaler()

# Compute mean and std from TRAINING data and scale X_train
X_train_scaled = scaler.fit_transform(X_train)

# Transform TEST data using parameters learned from training
X_test_scaled = scaler.transform(X_test)

print("✅ Feature scaling completed successfully!")
print("Example sample BEFORE scaling :", X_train.iloc[0].values[:4])
print("Example sample AFTER scaling  :", X_train_scaled[0][:4].round(3))
```

# Step 7 — Model Selection & Training

### 💡 *How Logistic Regression Works:*
The model computes a weighted linear combination of input features:
$$z = w_1 x_1 + w_2 x_2 + \dots + w_n x_n + b$$

Then passes $z$ through the **Sigmoid Activation Function** to yield a valid probability $P(y=1|x)$:
$$P(y=1) = \frac{1}{1 + e^{-z}}$$

```python
# ================================================================
# MODEL INITIALIZATION & TRAINING
# ================================================================
model = LogisticRegression(max_iter=1000, random_state=42)

# Fit learns weights (coef_) and bias (intercept_)
model.fit(X_train_scaled, y_train)

print("✅ Model training completed!")
print(f"Intercept (b) : {model.intercept_[0]:.3f}")
print("\nLearned Feature Coefficients (Weights):")

coef_df = pd.DataFrame({
    'Feature': X.columns,
    'Weight': model.coef_[0]
}).sort_values(by='Weight', key=abs, ascending=False)

for _, row in coef_df.iterrows():
    direction = "↑ Increases Diabetes Risk" if row['Weight'] > 0 else "↓ Decreases Diabetes Risk"
    print(f"  {row['Feature']:25s} -> {row['Weight']:>7.3f}  ({direction})")
```

# Step 8 — Prediction

The trained model provides two prediction methods:
1. `.predict(X_test)` $\rightarrow$ Discrete predicted class ($0$ or $1$) using default $0.5$ threshold.
2. `.predict_proba(X_test)` $\rightarrow$ Continuous probability estimates for each class $[P(y=0), P(y=1)]$.
   - Particularly vital in medicine where risk assessment thresholds may be adjusted.

```python
# ================================================================
# PREDICTIONS ON UNSEEN TEST DATA
# ================================================================
y_pred = model.predict(X_test_scaled)
y_proba = model.predict_proba(X_test_scaled)[:, 1]   # probability of class 1 (Diabetic)

# Sample comparison table
sample_results = pd.DataFrame({
    'Actual Label': y_test.values[:10],
    'Predicted Label': y_pred[:10],
    'Predicted Risk P(1)': y_proba[:10].round(3)
})
sample_results['Correct?'] = sample_results['Actual Label'] == sample_results['Predicted Label']
sample_results
```

# Step 9 — Model Evaluation

Classification evaluation requires a comprehensive set of diagnostic metrics:

| Metric | Formula | Medical Context Interpretation |
|---|---|---|
| **Accuracy** | $\frac{TP+TN}{\text{Total}}$ | Overall percentage of correct diagnostic predictions. |
| **Precision** | $\frac{TP}{TP+FP}$ | Out of all predicted diabetics, how many actually have diabetes? (Minimizes false alarms). |
| **Recall (Sensitivity)** | $\frac{TP}{TP+FN}$ | Out of all actual diabetic patients, how many did the model catch? (**Critical in healthcare!**). |
| **F1 Score** | $2 \cdot \frac{\text{Precision} \cdot \text{Recall}}{\text{Precision} + \text{Recall}}$ | Harmonic mean balancing Precision and Recall. |
| **ROC-AUC** | Area Under ROC Curve | Overall diagnostic capability across all decision thresholds ($0.5 = \text{Random}, 1.0 = \text{Perfect}$). |

```python
# ================================================================
# PERFORMANCE METRICS CALCULATION
# ================================================================
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
auc = roc_auc_score(y_test, y_proba)

print("=" * 50)
print("      EVALUATION METRICS REPORT CARD")
print("=" * 50)
print(f"  Accuracy       : {acc:.4f} ({acc*100:.1f}%)")
print(f"  Precision      : {prec:.4f} ({prec*100:.1f}%)")
print(f"  Recall (Sens.) : {rec:.4f} ({rec*100:.1f}%)")
print(f"  F1 Score       : {f1:.4f}")
print(f"  ROC-AUC Score  : {auc:.4f}")
print("=" * 50)

print("\n📋 Classification Report Breakdown:")
print(classification_report(y_test, y_pred, target_names=['Non-Diabetic (0)', 'Diabetic (1)']))
```

```python
# Confusion Matrix Visualization
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Pred Non-Diabetic', 'Pred Diabetic'],
            yticklabels=['Actual Non-Diabetic', 'Actual Diabetic'])
plt.title("Confusion Matrix Heatmap")
plt.ylabel("Actual")
plt.xlabel("Predicted")
plt.show()

tn, fp, fn, tp = cm.ravel()
print(f"✅ True Negatives  (TN) : {tn:2d}  -> Correctly identified healthy patients")
print(f"❌ False Positives (FP) : {fp:2d}  -> Healthy patients flagged as diabetic (False Alarm)")
print(f"❌ False Negatives (FN) : {fn:2d}  -> Diabetic patients MISSED (Dangerous error!)")
print(f"✅ True Positives  (TP) : {tp:2d}  -> Correctly identified diabetic patients")
```

```python
# Receiver Operating Characteristic (ROC) Curve
fpr, tpr, thresholds = roc_curve(y_test, y_proba)

plt.figure(figsize=(7, 6))
plt.plot(fpr, tpr, color='#2980b9', lw=2.5, label=f'Logistic Regression (AUC = {auc:.3f})')
plt.plot([0, 1], [0, 1], color='#7f8c8d', lw=1.5, linestyle='--', label='Random Guess Baseline (AUC = 0.500)')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate (1 - Specificity)')
plt.ylabel('True Positive Rate (Recall / Sensitivity)')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc='lower right')
plt.show()
```

# Step 10 — Hyperparameter Tuning

Hyperparameters are architectural settings configured prior to model training.

For **Logistic Regression**, key hyperparameters include:
- **`C`** : Inverse of Regularization strength. Smaller values $\rightarrow$ stronger regularization (simpler model); larger values $\rightarrow$ weaker regularization.
- **`penalty`** : Norm used in penalization (`'l1'` Lasso vs `'l2'` Ridge).
- **`solver`** : Algorithm used in the optimization problem (`'liblinear'` supports both L1 and L2 penalties).

```python
# ================================================================
# HYPERPARAMETER TUNING USING GridSearchCV
# ================================================================
param_grid = {
    'C': [0.001, 0.01, 0.1, 1.0, 10.0, 100.0],
    'penalty': ['l1', 'l2'],
    'solver': ['liblinear']
}

grid = GridSearchCV(
    LogisticRegression(max_iter=1000, random_state=42),
    param_grid,
    cv=5,                 # 5-fold cross-validation
    scoring='f1',          # optimize for balanced F1 score
    n_jobs=-1
)

grid.fit(X_train_scaled, y_train)

print("✅ GridSearchCV Optimization Complete!")
print(f"Best Hyperparameters : {grid.best_params_}")
print(f"Best Cross-Val F1    : {grid.best_score_:.4f}")

# Evaluate tuned model on the unseen test set
best_model = grid.best_estimator_
y_pred_tuned = best_model.predict(X_test_scaled)
y_proba_tuned = best_model.predict_proba(X_test_scaled)[:, 1]

acc_tuned = accuracy_score(y_test, y_pred_tuned)
prec_tuned = precision_score(y_test, y_pred_tuned)
rec_tuned = recall_score(y_test, y_pred_tuned)
f1_tuned = f1_score(y_test, y_pred_tuned)
auc_tuned = roc_auc_score(y_test, y_proba_tuned)
```

```python
# ================================================================
# FINAL MODEL COMPARISON: Baseline vs Tuned
# ================================================================
comparison_df = pd.DataFrame({
    'Model': ['Baseline Logistic Regression', 'Tuned Logistic Regression (GridSearch)'],
    'Accuracy': [acc, acc_tuned],
    'Precision': [prec, prec_tuned],
    'Recall': [rec, rec_tuned],
    'F1 Score': [f1, f1_tuned],
    'ROC-AUC': [auc, auc_tuned]
})

print("=" * 80)
print("FINAL MODEL COMPARISON TABLE")
print("=" * 80)
comparison_df.round(4)
```

## Step 11 — Save the Final Model & Scaler

We save both the **trained model** and the **scaler** using `joblib` so future patient records can be scaled and predicted identically without retraining.

```python
# ================================================================
# SERIALIZE & SAVE TO DISK
# ================================================================
final_model = best_model

joblib.dump(final_model, "logistic_regression_diabetes_model.pkl")
joblib.dump(scaler, "diabetes_feature_scaler.pkl")

print("✅ Model saved as 'logistic_regression_diabetes_model.pkl'")
print("✅ Scaler saved as 'diabetes_feature_scaler.pkl'")
```

```python
# ================================================================
# BONUS: Load the saved model and score a brand-new patient
# ================================================================
loaded_model = joblib.load("logistic_regression_diabetes_model.pkl")
loaded_scaler = joblib.load("diabetes_feature_scaler.pkl")

# Create hypothetical patient diagnostic record
new_patient_data = pd.DataFrame([{
    'Pregnancies': 3,
    'Glucose': 145.0,
    'BloodPressure': 78.0,
    'SkinThickness': 28.0,
    'Insulin': 130.0,
    'BMI': 32.4,
    'DiabetesPedigreeFunction': 0.65,
    'Age': 42
}])[X.columns]   # maintain exact column alignment

# Scale using training parameters
new_patient_scaled = loaded_scaler.transform(new_patient_data)

# Predict diagnostic outcome
pred_class = loaded_model.predict(new_patient_scaled)[0]
pred_prob = loaded_model.predict_proba(new_patient_scaled)[0][1]

print("=" * 45)
print("   NEW PATIENT DIAGNOSTIC PREDICTION")
print("=" * 45)
print(f"Predicted Class : {'🔴 Diabetic (1)' if pred_class == 1 else '🟢 Non-Diabetic (0)'}")
print(f"Diabetes Risk   : {pred_prob * 100:.2f}%")
print("=" * 45)
```

### 🏆 Golden Rules to Always Remember
1. **Always use `stratify=y`** when splitting data for classification.
2. **`fit_transform()` only on Training data**, and **only `transform()` on Test data** to prevent data leakage.
3. **In medical data, check for impossible zeros** (Glucose, Blood Pressure, BMI cannot be zero).
4. **Accuracy alone is misleading** — always examine Recall, Precision, and the Confusion Matrix.
5. **High Recall is essential in healthcare**: A False Negative (missing a sick patient) is far more dangerous than a False Positive (a false alarm).
